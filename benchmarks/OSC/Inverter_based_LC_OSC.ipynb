```julia
# 0. Set Python Environment
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"

using Printf
using StaticArrays
using PythonCall
using Random
using JLD2
using Interpolations
using Statistics
using Base.Threads

# ==============================================================================
# 1. Call the Surrogate definitions and the simulator Engine
# ==============================================================================
# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim_Finalized.jl")


# ==============================================================================
# 2. LTSPICE VERIFICATION GROUND TRUTH 
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, subprocess, numpy as np
from PyLTSpice import RawRead

def run_ltspice_lc_tran(Lp, Ln, mp, mn, mcp, mcn, CL, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    
    netlist = f'''* Cascode-Limited Inverter-Based LC VCO
.inc cmos_tt.txt
.options plotwinsize=0 method=trap
.tran 0 50n 0 10p uic

V0 vdd 0 3

* LEFT BRANCH
M4 ndP_L VY vdd vdd pmos_3p3 L={Lp} W=500n m={mp}
M2 VX 0 ndP_L vdd pmos_3p3 L={Lp} W=500n m={mcp}

M3 VX vdd ndN_L 0 nmos_3p3 L={Ln} W=500n m={mcn}
M1 ndN_L VY 0 0 nmos_3p3 L={Ln} W=500n m={mn}

* RIGHT BRANCH
M5 ndP_R VX vdd vdd pmos_3p3 L={Lp} W=500n m={mp}
M6 VY 0 ndP_R vdd pmos_3p3 L={Lp} W=500n m={mcp}

M7 VY vdd ndN_R 0 nmos_3p3 L={Ln} W=500n m={mcn}
M0 ndN_R VX 0 0 nmos_3p3 L={Ln} W=500n m={mn}

* LC TANK
L2 VX n_L2 5n
R0 n_L2 n_R0 5
L0 n_R0 n_L0 5n
R1 n_L0 VY 5

C2 VX n_C 200f
C0 n_C VY 200f

* Load Capacitors
C4 VX 0 {CL}
C3 VY 0 {CL}

.ic V(VX)=3

.model NMOS NMOS
.model PMOS PMOS
.lib {os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')}

.meas tran t_rise1 TRIG v(VX) VAL=2 RISE=50
.meas tran t_rise2 TRIG v(VX) VAL=2 RISE=51
.meas tran f_osc param abs(1/(t_rise2-t_rise1))
.end
'''
    sp_path = os.path.join(BASE_DIR, "tran_test_lc.sp")
    raw_path = os.path.join(BASE_DIR, "tran_test_lc.raw")
    log_path = os.path.join(BASE_DIR, "tran_test_lc.log")
    
    if os.path.exists(raw_path):
        os.remove(raw_path)
        
    with open(sp_path, 'w', encoding='utf-8') as f: f.write(netlist)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp_path], cwd=BASE_DIR, capture_output=True)
    raw = RawRead(raw_path)
    t = raw.get_trace("time").get_wave(0)
    vx = raw.get_trace("V(vx)").get_wave(0)
    vy = raw.get_trace("V(vy)").get_wave(0)
    
    f_osc = 0.0
    if os.path.exists(log_path):
        with open(log_path, 'r', encoding='utf-8') as f:
            for line in f:
                if line.strip().startswith('f_osc:'):
                    try: f_osc = float(line.split('=')[1].strip())
                    except: pass
    return t, vx, vy, f_osc
""", g)
run_ltspice_lc_tran = g["run_ltspice_lc_tran"]

# ==============================================================================
# 3. CIRCUIT DEFINITIONS & 300-BATCH SETUP
# ==============================================================================
struct CircuitParams
    Lp::Float64
    Ln::Float64
    mp::Float64
    mcp::Float64
    mn::Float64
    mcn::Float64
    CL::Float64
end

N = 300
CL_arr = range(1e-12, 2e-12, length=N)

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    circuits[i] = CircuitParams(0.3e-6, 0.3e-6, 70.0, 105.0, 20.0, 30.0, CL_arr[i])
end

println("\n====================================================================")
println(" TRANSIENT BATCH EVALUATION: $(N) CIRCUITS (INVERTER-BASED LC VCO)")
println("====================================================================")

sim_tran = JuliaGSim(surrogate_model)
add_vsource!(sim_tran, "V0", "vdd", "0", 3.0)

# LEFT BRANCH
add_pmos!(sim_tran, "M4", d="ndP_L", g="VY", s="vdd", b="vdd", M=:mp, L=:Lp)
add_pmos!(sim_tran, "M2", d="VX", g="0", s="ndP_L", b="vdd", M=:mcp, L=:Lp)

add_nmos!(sim_tran, "M3", d="VX", g="vdd", s="ndN_L", b="0", M=:mcn, L=:Ln)
add_nmos!(sim_tran, "M1", d="ndN_L", g="VY", s="0", b="0", M=:mn, L=:Ln)

# RIGHT BRANCH
add_pmos!(sim_tran, "M5", d="ndP_R", g="VX", s="vdd", b="vdd", M=:mp, L=:Lp)
add_pmos!(sim_tran, "M6", d="VY", g="0", s="ndP_R", b="vdd", M=:mcp, L=:Lp)

add_nmos!(sim_tran, "M7", d="VY", g="vdd", s="ndN_R", b="0", M=:mcn, L=:Ln)
add_nmos!(sim_tran, "M0", d="ndN_R", g="VX", s="0", b="0", M=:mn, L=:Ln)

# LC TANK
add_ind!(sim_tran, "L2", "VX", "n_L2", 5e-9)
add_res!(sim_tran, "R0", "n_L2", "n_R0", 5.0)
add_ind!(sim_tran, "L0", "n_R0", "n_L0", 5e-9)
add_res!(sim_tran, "R1", "n_L0", "VY", 5.0)

add_cap!(sim_tran, "C2", "VX", "n_C", 200e-15)
add_cap!(sim_tran, "C0", "n_C", "VY", 200e-15)

# Load Capacitors
add_cap!(sim_tran, "C4", "VX", "0", :CL; ic=3.0)
add_cap!(sim_tran, "C3", "VY", "0", :CL)

compile!(sim_tran)

BASE_DIR = pwd()

# ==============================================================================
# 4. RUN JULIAGSIM (ALWAYS RUNS FRESH & OVERWRITES CACHE)
# ==============================================================================
println("\n>>> Running $(N)-Batch Transient Solves on $(nthreads()) Julia Threads (Trapezoidal Method)...")

t_stop = 50e-9
dt_init = 10e-12
dt_max = 10e-12 

results_x_hist = Vector{Matrix{Float64}}(undef, N)
results_t_hist = Vector{Vector{Float64}}(undef, N)

start_time = time()
@threads for i in 1:N
    t_hist, x_hist = solve_tran(sim_tran, circuits[i], t_stop, dt_init, uic=true, dt_max=dt_max, method=:trapezoid)
    results_x_hist[i] = x_hist
    results_t_hist[i] = t_hist
end
total_time = time() - start_time

println("Julia $(N)-Circuit Transient Total Time: $(round(total_time, digits=4)) s")

# Save/overwrite fresh Julia simulation results to disk
julia_cache_file = joinpath(BASE_DIR, "julia_lc_osc_cache_$(N).jld2")
@save julia_cache_file results_t_hist results_x_hist total_time
println("    Saved fresh Julia results to: $julia_cache_file")

idx_vx = findfirst(==("VX"), sim_tran.unknown_nodes)
idx_vy = findfirst(==("VY"), sim_tran.unknown_nodes)

# ==============================================================================
# 5. RUN LTSPICE BASELINE (ALWAYS RUNS FRESH & OVERWRITES CACHE)
# ==============================================================================
println("\n>>> Running LTspice baseline (Sequential)...")

spice_times = Vector{Any}(undef, N)
spice_vxs   = Vector{Any}(undef, N)
spice_vys   = Vector{Any}(undef, N)
spice_freqs = zeros(N)

start_spice = time()
for i in 1:N
    t_lt, vx_lt, vy_lt, f_osc_lt = run_ltspice_lc_tran(
        circuits[i].Lp, circuits[i].Ln, circuits[i].mp, circuits[i].mn, 
        circuits[i].mcp, circuits[i].mcn, circuits[i].CL, BASE_DIR
    )
    spice_times[i] = pyconvert(Array, t_lt)
    spice_vxs[i]   = pyconvert(Array, vx_lt)
    spice_vys[i]   = pyconvert(Array, vy_lt)
    spice_freqs[i] = pyconvert(Float64, f_osc_lt)
    if i % 50 == 0; println("    LTspice progress: $i/$N circuits done..."); end
end
spice_time = time() - start_spice

println("LTspice Total Time: $(round(spice_time, digits=4)) s")

# Save/overwrite fresh LTspice simulation results to disk
spice_cache_file = joinpath(BASE_DIR, "ltspice_lc_osc_cache_$(N).jld2")
@save spice_cache_file spice_times spice_vxs spice_vys spice_freqs spice_time
println("    Saved fresh LTspice results to: $spice_cache_file")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 TRANSIENT PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")

# ==============================================================================
# 6. FREQUENCY EXTRACTION & FULL 300-CIRCUIT COMPARISON TABLE
# ==============================================================================
jl_freqs = zeros(N)
for i in 1:N
    times = results_t_hist[i]
    vx_arr = results_x_hist[i][idx_vx, :]
    
    t_rise1, t_rise2 = 0.0, 0.0
    rise_count = 0
    for j in 2:length(times)
        if vx_arr[j-1] < 2.0 && vx_arr[j] >= 2.0
            t_cross = times[j-1] + (2.0 - vx_arr[j-1]) * (times[j] - times[j-1]) / (vx_arr[j] - vx_arr[j-1])
            rise_count += 1
            if rise_count == 50
                t_rise1 = t_cross
            elseif rise_count == 51
                t_rise2 = t_cross
                break
            end
        end
    end
    jl_freqs[i] = (rise_count >= 51) ? abs(1.0 / (t_rise2 - t_rise1)) : 0.0
end

println("==========================================================================================")
println(" DETAILED CIRCUIT RESULTS (FREQUENCY COMPARISON)")
println("==========================================================================================\n")
@printf("%-10s | %-12s | %-15s | %-15s | %-15s\n", "Circuit", "CL (pF)", "Julia Freq", "LTspice Freq", "Error (%)")
println(repeat("-", 75))

freq_errors = zeros(N)
for i in 1:N
    err = spice_freqs[i] > 0 ? abs(jl_freqs[i] - spice_freqs[i]) / spice_freqs[i] * 100 : 0.0
    freq_errors[i] = err
    @printf("%-10d | %-12.3f | %-11.2f MHz | %-11.2f MHz | %-11.2f %%\n", i, circuits[i].CL * 1e12, jl_freqs[i] / 1e6, spice_freqs[i] / 1e6, err)
end
println()

println("======================================================================")
println(" OVERALL ACCURACY STATISTICAL SUMMARY ($(N) CIRCUITS)")
println("======================================================================")
@printf(" Max Frequency Error     : %.2f %%\n", maximum(freq_errors))
@printf(" Mean Frequency Error    : %.2f %%\n", mean(freq_errors))
@printf(" Min Oscillation Freq    : %.2f MHz\n", minimum(jl_freqs)/1e6)
@printf(" Max Oscillation Freq    : %.2f MHz\n", maximum(jl_freqs)/1e6)
println("======================================================================\n")

# ==============================================================================
# 7. VCO TUNING CHARACTERISTIC: fosc vs. CL (RED for Julia, BLUE for LTspice)
# ==============================================================================
plt = pyimport("matplotlib.pyplot")

fig_tune = plt.figure(figsize=(10, 5))
CL_vals_pF = [circuits[i].CL * 1e12 for i in 1:N]

plt.plot(CL_vals_pF, jl_freqs ./ 1e6, color="red", linewidth=2.5, label="JuliaGSim")
plt.plot(CL_vals_pF, spice_freqs ./ 1e6, color="blue", linestyle="--", linewidth=2.0, label="LTspice")
plt.title("Inverter-Based LC VCO Tuning: Oscillation Frequency vs. Load Capacitance ($(N) Circuits)", fontsize=12)
plt.xlabel("Load Capacitance CL (pF)", fontsize=11)
plt.ylabel("Oscillation Frequency (MHz)", fontsize=11)
plt.legend(loc="upper right", fontsize=11)
plt.grid(true, which="both", ls=":")
plt.tight_layout()

buf_tune = pyimport("io").BytesIO()
fig_tune.savefig(buf_tune, format="png", bbox_inches="tight")
display(MIME("image/png"), pyconvert(Vector{UInt8}, buf_tune.getvalue()))
plt.close(fig_tune)
println()

# ==============================================================================
# 8. WAVEFORM PLOTS: TOP 20 CIRCUITS WITH HIGHEST MISMATCH (40 IMAGES TOTAL)
# ==============================================================================
worst_order = sortperm(freq_errors, rev=true)
worst_20 = worst_order[1:min(20, N)]

println("======================================================================")
println(" WAVEFORM PLOTS: TOP 20 CIRCUITS WITH HIGHEST FREQUENCY MISMATCH")
println(" (2 SEPARATE FIGURES PER CIRCUIT: VX, VY -> TOTAL 40 FIGURES)")
println("======================================================================\n")

for (rank, i) in enumerate(worst_20)
    times_jl_arr = results_t_hist[i]
    vx_arr = results_x_hist[i][idx_vx, :]
    vy_arr = results_x_hist[i][idx_vy, :]

    println(repeat("=", 75))
    @printf(" Worst Case #%d | Circuit %d | CL = %.3f pF | Freq Err = %.2f %%\n", rank, i, circuits[i].CL * 1e12, freq_errors[i])
    @printf(" JuliaGSim Freq: %.2f MHz | LTspice Freq: %.2f MHz\n", jl_freqs[i]/1e6, spice_freqs[i]/1e6)
    println(repeat("=", 75))

    # PLOT 1: VX
    fig1 = plt.figure(figsize=(10, 3))
    plt.plot(times_jl_arr .* 1e9, vx_arr, color="red", linewidth=2, label="JuliaGSim Vx")
    plt.plot(spice_times[i] .* 1e9, spice_vxs[i], color="blue", linestyle="--", label="LTspice Vx")
    plt.title("Circuit $i (CL=$(round(circuits[i].CL*1e12, digits=2))pF) - Node VX")
    plt.xlabel("Time (ns)")
    plt.ylabel("Voltage (V)")
    plt.legend(loc="upper right")
    plt.grid(true, which="both", ls=":")
    plt.tight_layout()
    buf1 = pyimport("io").BytesIO()
    fig1.savefig(buf1, format="png", bbox_inches="tight")
    display(MIME("image/png"), pyconvert(Vector{UInt8}, buf1.getvalue()))
    plt.close(fig1)

    # PLOT 2: VY
    fig2 = plt.figure(figsize=(10, 3))
    plt.plot(times_jl_arr .* 1e9, vy_arr, color="green", linewidth=2, label="JuliaGSim Vy")
    plt.plot(spice_times[i] .* 1e9, spice_vys[i], color="black", linestyle="--", label="LTspice Vy")
    plt.title("Circuit $i (CL=$(round(circuits[i].CL*1e12, digits=2))pF) - Node VY")
    plt.xlabel("Time (ns)")
    plt.ylabel("Voltage (V)")
    plt.legend(loc="upper right")
    plt.grid(true, which="both", ls=":")
    plt.tight_layout()
    buf2 = pyimport("io").BytesIO()
    fig2.savefig(buf2, format="png", bbox_inches="tight")
    display(MIME("image/png"), pyconvert(Vector{UInt8}, buf2.getvalue()))
    plt.close(fig2)

    println(repeat("-", 60))
end
```

```julia
# 0. Set Python Environment
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"

using Printf
using StaticArrays
using PythonCall
using Random
using JLD2
using Interpolations
using Statistics
using Base.Threads

# ==============================================================================
# 1. Call the Surrogate definitions and the simulator Engine
# ==============================================================================
# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim_Finalized.jl")


# ==============================================================================
# 2. LTSPICE VERIFICATION GROUND TRUTH 
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, subprocess, numpy as np
from PyLTSpice import RawRead

def run_ltspice_lc_tran(Lp, Ln, mp, mn, mcp, mcn, CL, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    
    netlist = f'''* Cascode-Limited Inverter-Based LC VCO
.inc cmos_tt.txt
.options plotwinsize=0 method=trap
.tran 0 50n 0 10p uic

V0 vdd 0 3

* LEFT BRANCH
M4 ndP_L VY vdd vdd pmos_3p3 L={Lp} W=500n m={mp}
M2 VX 0 ndP_L vdd pmos_3p3 L={Lp} W=500n m={mcp}

M3 VX vdd ndN_L 0 nmos_3p3 L={Ln} W=500n m={mcn}
M1 ndN_L VY 0 0 nmos_3p3 L={Ln} W=500n m={mn}

* RIGHT BRANCH
M5 ndP_R VX vdd vdd pmos_3p3 L={Lp} W=500n m={mp}
M6 VY 0 ndP_R vdd pmos_3p3 L={Lp} W=500n m={mcp}

M7 VY vdd ndN_R 0 nmos_3p3 L={Ln} W=500n m={mcn}
M0 ndN_R VX 0 0 nmos_3p3 L={Ln} W=500n m={mn}

* LC TANK
L2 VX n_L2 5n
R0 n_L2 n_R0 5
L0 n_R0 n_L0 5n
R1 n_L0 VY 5

C2 VX n_C 200f
C0 n_C VY 200f

* Load Capacitors
C4 VX 0 {CL}
C3 VY 0 {CL}

.ic V(VX)=3

.model NMOS NMOS
.model PMOS PMOS
.lib {os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')}

.meas tran t_rise1 TRIG v(VX) VAL=2 RISE=50
.meas tran t_rise2 TRIG v(VX) VAL=2 RISE=51
.meas tran f_osc param abs(1/(t_rise2-t_rise1))
.end
'''
    sp_path = os.path.join(BASE_DIR, "tran_test_lc.sp")
    raw_path = os.path.join(BASE_DIR, "tran_test_lc.raw")
    log_path = os.path.join(BASE_DIR, "tran_test_lc.log")
    
    if os.path.exists(raw_path):
        os.remove(raw_path)
        
    with open(sp_path, 'w', encoding='utf-8') as f: f.write(netlist)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp_path], cwd=BASE_DIR, capture_output=True)
    raw = RawRead(raw_path)
    t = raw.get_trace("time").get_wave(0)
    vx = raw.get_trace("V(vx)").get_wave(0)
    vy = raw.get_trace("V(vy)").get_wave(0)
    
    f_osc = 0.0
    if os.path.exists(log_path):
        with open(log_path, 'r', encoding='utf-8') as f:
            for line in f:
                if line.strip().startswith('f_osc:'):
                    try: f_osc = float(line.split('=')[1].strip())
                    except: pass
    return t, vx, vy, f_osc
""", g)
run_ltspice_lc_tran = g["run_ltspice_lc_tran"]

# ==============================================================================
# 3. CIRCUIT DEFINITIONS & 300-BATCH SETUP
# ==============================================================================
struct CircuitParams
    Lp::Float64
    Ln::Float64
    mp::Float64
    mcp::Float64
    mn::Float64
    mcn::Float64
    CL::Float64
end

N = 300
CL_arr = range(1e-12, 2e-12, length=N)

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    circuits[i] = CircuitParams(0.3e-6, 0.3e-6, 70.0, 105.0, 20.0, 30.0, CL_arr[i])
end

println("\n====================================================================")
println(" TRANSIENT BATCH EVALUATION: $(N) CIRCUITS (INVERTER-BASED LC VCO)")
println("====================================================================")

sim_tran = JuliaGSim(surrogate_model)
add_vsource!(sim_tran, "V0", "vdd", "0", 3.0)

# LEFT BRANCH
add_pmos!(sim_tran, "M4", d="ndP_L", g="VY", s="vdd", b="vdd", M=:mp, L=:Lp)
add_pmos!(sim_tran, "M2", d="VX", g="0", s="ndP_L", b="vdd", M=:mcp, L=:Lp)

add_nmos!(sim_tran, "M3", d="VX", g="vdd", s="ndN_L", b="0", M=:mcn, L=:Ln)
add_nmos!(sim_tran, "M1", d="ndN_L", g="VY", s="0", b="0", M=:mn, L=:Ln)

# RIGHT BRANCH
add_pmos!(sim_tran, "M5", d="ndP_R", g="VX", s="vdd", b="vdd", M=:mp, L=:Lp)
add_pmos!(sim_tran, "M6", d="VY", g="0", s="ndP_R", b="vdd", M=:mcp, L=:Lp)

add_nmos!(sim_tran, "M7", d="VY", g="vdd", s="ndN_R", b="0", M=:mcn, L=:Ln)
add_nmos!(sim_tran, "M0", d="ndN_R", g="VX", s="0", b="0", M=:mn, L=:Ln)

# LC TANK
add_ind!(sim_tran, "L2", "VX", "n_L2", 5e-9)
add_res!(sim_tran, "R0", "n_L2", "n_R0", 5.0)
add_ind!(sim_tran, "L0", "n_R0", "n_L0", 5e-9)
add_res!(sim_tran, "R1", "n_L0", "VY", 5.0)

add_cap!(sim_tran, "C2", "VX", "n_C", 200e-15)
add_cap!(sim_tran, "C0", "n_C", "VY", 200e-15)

# Load Capacitors
add_cap!(sim_tran, "C4", "VX", "0", :CL; ic=3.0)
add_cap!(sim_tran, "C3", "VY", "0", :CL)

compile!(sim_tran)

BASE_DIR = pwd()

# ==============================================================================
# 4. RUN JULIAGSIM (ALWAYS RUNS FRESH & OVERWRITES CACHE)
# ==============================================================================
println("\n>>> Running $(N)-Batch Transient Solves on $(nthreads()) Julia Threads (Gear Method)...")

t_stop = 50e-9
dt_init = 10e-12
dt_max = 10e-12 

results_x_hist = Vector{Matrix{Float64}}(undef, N)
results_t_hist = Vector{Vector{Float64}}(undef, N)

start_time = time()
@threads for i in 1:N
    t_hist, x_hist = solve_tran(sim_tran, circuits[i], t_stop, dt_init, uic=true, dt_max=dt_max, method=:gear)
    results_x_hist[i] = x_hist
    results_t_hist[i] = t_hist
end
total_time = time() - start_time

println("Julia $(N)-Circuit Transient Total Time: $(round(total_time, digits=4)) s")

# Save/overwrite fresh Julia simulation results to disk
julia_cache_file = joinpath(BASE_DIR, "julia_lc_osc_cache_$(N).jld2")
@save julia_cache_file results_t_hist results_x_hist total_time
println("    Saved fresh Julia results to: $julia_cache_file")

idx_vx = findfirst(==("VX"), sim_tran.unknown_nodes)
idx_vy = findfirst(==("VY"), sim_tran.unknown_nodes)

# ==============================================================================
# 5. RUN LTSPICE BASELINE (ALWAYS RUNS FRESH & OVERWRITES CACHE)
# ==============================================================================
println("\n>>> Running LTspice baseline (Sequential)...")

spice_times = Vector{Any}(undef, N)
spice_vxs   = Vector{Any}(undef, N)
spice_vys   = Vector{Any}(undef, N)
spice_freqs = zeros(N)

start_spice = time()
for i in 1:N
    t_lt, vx_lt, vy_lt, f_osc_lt = run_ltspice_lc_tran(
        circuits[i].Lp, circuits[i].Ln, circuits[i].mp, circuits[i].mn, 
        circuits[i].mcp, circuits[i].mcn, circuits[i].CL, BASE_DIR
    )
    spice_times[i] = pyconvert(Array, t_lt)
    spice_vxs[i]   = pyconvert(Array, vx_lt)
    spice_vys[i]   = pyconvert(Array, vy_lt)
    spice_freqs[i] = pyconvert(Float64, f_osc_lt)
    if i % 50 == 0; println("    LTspice progress: $i/$N circuits done..."); end
end
spice_time = time() - start_spice

println("LTspice Total Time: $(round(spice_time, digits=4)) s")

# Save/overwrite fresh LTspice simulation results to disk
spice_cache_file = joinpath(BASE_DIR, "ltspice_lc_osc_cache_$(N).jld2")
@save spice_cache_file spice_times spice_vxs spice_vys spice_freqs spice_time
println("    Saved fresh LTspice results to: $spice_cache_file")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 TRANSIENT PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")

# ==============================================================================
# 6. FREQUENCY EXTRACTION & FULL 300-CIRCUIT COMPARISON TABLE
# ==============================================================================
jl_freqs = zeros(N)
for i in 1:N
    times = results_t_hist[i]
    vx_arr = results_x_hist[i][idx_vx, :]
    
    t_rise1, t_rise2 = 0.0, 0.0
    rise_count = 0
    for j in 2:length(times)
        if vx_arr[j-1] < 2.0 && vx_arr[j] >= 2.0
            t_cross = times[j-1] + (2.0 - vx_arr[j-1]) * (times[j] - times[j-1]) / (vx_arr[j] - vx_arr[j-1])
            rise_count += 1
            if rise_count == 50
                t_rise1 = t_cross
            elseif rise_count == 51
                t_rise2 = t_cross
                break
            end
        end
    end
    jl_freqs[i] = (rise_count >= 51) ? abs(1.0 / (t_rise2 - t_rise1)) : 0.0
end

println("==========================================================================================")
println(" DETAILED CIRCUIT RESULTS (FREQUENCY COMPARISON)")
println("==========================================================================================\n")
@printf("%-10s | %-12s | %-15s | %-15s | %-15s\n", "Circuit", "CL (pF)", "Julia Freq", "LTspice Freq", "Error (%)")
println(repeat("-", 75))

freq_errors = zeros(N)
for i in 1:N
    err = spice_freqs[i] > 0 ? abs(jl_freqs[i] - spice_freqs[i]) / spice_freqs[i] * 100 : 0.0
    freq_errors[i] = err
    @printf("%-10d | %-12.3f | %-11.2f MHz | %-11.2f MHz | %-11.2f %%\n", i, circuits[i].CL * 1e12, jl_freqs[i] / 1e6, spice_freqs[i] / 1e6, err)
end
println()

println("======================================================================")
println(" OVERALL ACCURACY STATISTICAL SUMMARY ($(N) CIRCUITS)")
println("======================================================================")
@printf(" Max Frequency Error     : %.2f %%\n", maximum(freq_errors))
@printf(" Mean Frequency Error    : %.2f %%\n", mean(freq_errors))
@printf(" Min Oscillation Freq    : %.2f MHz\n", minimum(jl_freqs)/1e6)
@printf(" Max Oscillation Freq    : %.2f MHz\n", maximum(jl_freqs)/1e6)
println("======================================================================\n")

# ==============================================================================
# 7. VCO TUNING CHARACTERISTIC: fosc vs. CL (RED for Julia, BLUE for LTspice)
# ==============================================================================
plt = pyimport("matplotlib.pyplot")

fig_tune = plt.figure(figsize=(10, 5))
CL_vals_pF = [circuits[i].CL * 1e12 for i in 1:N]

plt.plot(CL_vals_pF, jl_freqs ./ 1e6, color="red", linewidth=2.5, label="JuliaGSim")
plt.plot(CL_vals_pF, spice_freqs ./ 1e6, color="blue", linestyle="--", linewidth=2.0, label="LTspice")
plt.title("Inverter-Based LC VCO Tuning: Oscillation Frequency vs. Load Capacitance ($(N) Circuits)", fontsize=12)
plt.xlabel("Load Capacitance CL (pF)", fontsize=11)
plt.ylabel("Oscillation Frequency (MHz)", fontsize=11)
plt.legend(loc="upper right", fontsize=11)
plt.grid(true, which="both", ls=":")
plt.tight_layout()

buf_tune = pyimport("io").BytesIO()
fig_tune.savefig(buf_tune, format="png", bbox_inches="tight")
display(MIME("image/png"), pyconvert(Vector{UInt8}, buf_tune.getvalue()))
plt.close(fig_tune)
println()

# ==============================================================================
# 8. WAVEFORM PLOTS: TOP 20 CIRCUITS WITH HIGHEST MISMATCH (40 IMAGES TOTAL)
# ==============================================================================
worst_order = sortperm(freq_errors, rev=true)
worst_20 = worst_order[1:min(20, N)]

println("======================================================================")
println(" WAVEFORM PLOTS: TOP 20 CIRCUITS WITH HIGHEST FREQUENCY MISMATCH")
println(" (2 SEPARATE FIGURES PER CIRCUIT: VX, VY -> TOTAL 40 FIGURES)")
println("======================================================================\n")

for (rank, i) in enumerate(worst_20)
    times_jl_arr = results_t_hist[i]
    vx_arr = results_x_hist[i][idx_vx, :]
    vy_arr = results_x_hist[i][idx_vy, :]

    println(repeat("=", 75))
    @printf(" Worst Case #%d | Circuit %d | CL = %.3f pF | Freq Err = %.2f %%\n", rank, i, circuits[i].CL * 1e12, freq_errors[i])
    @printf(" JuliaGSim Freq: %.2f MHz | LTspice Freq: %.2f MHz\n", jl_freqs[i]/1e6, spice_freqs[i]/1e6)
    println(repeat("=", 75))

    # PLOT 1: VX
    fig1 = plt.figure(figsize=(10, 3))
    plt.plot(times_jl_arr .* 1e9, vx_arr, color="red", linewidth=2, label="JuliaGSim Vx")
    plt.plot(spice_times[i] .* 1e9, spice_vxs[i], color="blue", linestyle="--", label="LTspice Vx")
    plt.title("Circuit $i (CL=$(round(circuits[i].CL*1e12, digits=2))pF) - Node VX")
    plt.xlabel("Time (ns)")
    plt.ylabel("Voltage (V)")
    plt.legend(loc="upper right")
    plt.grid(true, which="both", ls=":")
    plt.tight_layout()
    buf1 = pyimport("io").BytesIO()
    fig1.savefig(buf1, format="png", bbox_inches="tight")
    display(MIME("image/png"), pyconvert(Vector{UInt8}, buf1.getvalue()))
    plt.close(fig1)

    # PLOT 2: VY
    fig2 = plt.figure(figsize=(10, 3))
    plt.plot(times_jl_arr .* 1e9, vy_arr, color="green", linewidth=2, label="JuliaGSim Vy")
    plt.plot(spice_times[i] .* 1e9, spice_vys[i], color="black", linestyle="--", label="LTspice Vy")
    plt.title("Circuit $i (CL=$(round(circuits[i].CL*1e12, digits=2))pF) - Node VY")
    plt.xlabel("Time (ns)")
    plt.ylabel("Voltage (V)")
    plt.legend(loc="upper right")
    plt.grid(true, which="both", ls=":")
    plt.tight_layout()
    buf2 = pyimport("io").BytesIO()
    fig2.savefig(buf2, format="png", bbox_inches="tight")
    display(MIME("image/png"), pyconvert(Vector{UInt8}, buf2.getvalue()))
    plt.close(fig2)

    println(repeat("-", 60))
end
```
