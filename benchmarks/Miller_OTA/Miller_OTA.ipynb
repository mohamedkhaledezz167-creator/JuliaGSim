```julia
# ==============================================================================
# 1. Call the LUT definitions and the simulator Engine
# ==============================================================================
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"

using Random
using Printf
using StaticArrays
using PythonCall

# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim.jl")

# ==============================================================================
# 2.Helper to unwrap phase angles natively in Julia 
# ==============================================================================
function unwrap_phase_custom(p)
    p_out = copy(p)
    for j in 2:length(p_out)
        d = p_out[j] - p_out[j-1]
        p_out[j] -= 2pi * round(d / 2pi) 
    end
    return p_out
end

# ==============================================================================
# 3. LTSPICE VERIFICATION GROUND TRUTH (Self-Contained Dynamic Netlists)
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, subprocess, numpy as np
from PyLTSpice import RawRead

def evaluate_miller_ota_dc_voltages(md, Ld, mm, Lm, mss, Lss, mb, beta, Ib, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR, "Miller_P_DC.sp")
    rp = os.path.join(BASE_DIR, "Miller_P_DC.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')

    m6 = 2.0 * beta * mm
    m7 = beta * mss

    nl = f'''* PMOS-Input Miller OTA - DC Operating Point AutoGen
Vdd Vdd 0 3
Vicm vicm 0 {Vicm}
vip vip vicm 0
vin vin vicm 0

IB vb 0 {Ib}
M8 vb vb Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mb}

M5 tail vb Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mss}
M1 vx vin tail Vdd pmos_3p3 l={Ld} w=0.5u m={md}
M2 v1 vip tail Vdd pmos_3p3 l={Ld} w=0.5u m={md}
M3 vx vx 0 0 nmos_3p3 l={Lm} w=0.5u m={mm}
M4 v1 vx 0 0 nmos_3p3 l={Lm} w=0.5u m={mm}

M7 vout vb Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={m7}
M6 vout v1 0 0 nmos_3p3 l={Lm} w=0.5u m={m6}

Rz v1 v_comp 5000.0
Cc v_comp vout 1e-12
CL vout 0 1e-13

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.op
.end
'''
    with open(sp, 'w', encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp], cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(rp)
    tr = raw.get_trace_names()
    def gv(n):
        for t in tr:
            if t.lower() == n.lower(): return float(raw.get_trace(t).get_wave(0)[0])
        return 0.0
    return {k: gv(f"V({k})") for k in ['vb', 'tail', 'v1', 'vx', 'vout']}

def evaluate_miller_ota_ac_ltspice(md, Ld, mm, Lm, mss, Lss, mb, beta, Ib, Cc, Rz, CL, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR, "Miller_P_AC_TEST.sp")
    rp = os.path.join(BASE_DIR, "Miller_P_AC_TEST.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')

    nl = f'''* PMOS-Input Miller OTA - AC Testbench AutoGen
Vdd Vdd 0 3
vicm_dm vcm_dm 0 {Vicm}
vinp_dm vip_dm vcm_dm AC 0.5
vinn_dm vin_dm vcm_dm AC -0.5

IB_dm vb_dm 0 {Ib}
M8_dm vb_dm vb_dm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mb} ad=0 as=0 pd=0 ps=0
M5_dm tail_dm vb_dm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mss} ad=0 as=0 pd=0 ps=0
M1_dm vx_dm vin_dm tail_dm Vdd pmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M2_dm v1_dm vip_dm tail_dm Vdd pmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M3_dm vx_dm vx_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M4_dm v1_dm vx_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M7_dm vout_dm vb_dm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mss*beta} ad=0 as=0 pd=0 ps=0
M6_dm vout_dm v1_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm*2*beta} ad=0 as=0 pd=0 ps=0

Rz_dm v1_dm v_comp_dm {Rz}
Cc_dm v_comp_dm vout_dm {Cc}
CL_dm vout_dm 0 {CL}

vicm_ac vicm_ac 0 {Vicm} AC 1
IB_cm vb_cm 0 {Ib}
M8_cm vb_cm vb_cm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mb} ad=0 as=0 pd=0 ps=0
M5_cm tail_cm vb_cm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mss} ad=0 as=0 pd=0 ps=0
M1_cm vx_cm vicm_ac tail_cm Vdd pmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M2_cm v1_cm vicm_ac tail_cm Vdd pmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M3_cm vx_cm vx_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M4_cm v1_cm vx_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M7_cm vout_cm vb_cm Vdd Vdd pmos_3p3 l={Lss} w=0.5u m={mss*beta} ad=0 as=0 pd=0 ps=0
M6_cm vout_cm v1_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm*2*beta} ad=0 as=0 pd=0 ps=0

Rz_cm v1_cm v_comp_cm {Rz}
Cc_cm v_comp_cm vout_cm {Cc}
CL_cm vout_cm 0 {CL}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.ac dec 100 1 10G
.end
'''
    with open(sp, 'w', encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp], cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(rp)
    freqs = np.abs(raw.get_trace("frequency").get_wave(0))
    
    vout_dm_c = raw.get_trace("V(vout_dm)").get_wave(0)
    vout_dm = np.abs(vout_dm_c)
    vout_cm = np.abs(raw.get_trace("V(vout_cm)").get_wave(0))
    
    gain_dm_db = 20 * np.log10(vout_dm)
    gain_cm_db = 20 * np.log10(vout_cm)
    phase = np.unwrap(np.angle(vout_dm_c)) * 180 / np.pi
    
    AVD = np.max(gain_dm_db)
    bwi = np.where(gain_dm_db <= AVD - 3.0)[0]
    BW = freqs[bwi[0]] if len(bwi) else freqs[-1]
    ugf_i = np.argmin(np.abs(gain_dm_db))
    UGF = freqs[ugf_i]
    PM = 180.0 + phase[ugf_i]
    AVCM = gain_cm_db[0]
    
    return {'AVD': AVD, 'BW': BW, 'UGF': UGF, 'PM': PM, 'AVCM': AVCM, 'CMRR': AVD - AVCM,
            'freqs': freqs, 'mag': gain_dm_db, 'phase': phase}
""", g)
evaluate_miller_ota_dc = g["evaluate_miller_ota_dc_voltages"]
evaluate_miller_ota_ac = g["evaluate_miller_ota_ac_ltspice"]

# ==============================================================================
# 4. BUILDING THE DIFFERENTIAL AND COMMON MODE MILLER OTA
# ==============================================================================
VDD, Vicm, Ib = 3.0, 1.5, 1e-5
beta_val = 1.0
Cc, Rz, CL = 1e-12, 5000.0, 1e-13

sim_dm = JuliaGSim(surrogate_model)
add_vsource!(sim_dm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_dm, "V_VIP", "vip", "0", Vicm)
add_vsource!(sim_dm, "V_VIN", "vin", "0", Vicm)
add_isource!(sim_dm, "I_BIAS", "vb", "0", Ib)

add_vac_source!(sim_dm, "V_VIP_AC", "vip", "0", 0.5)
add_vac_source!(sim_dm, "V_VIN_AC", "vin", "0", -0.5)

add_pmos!(sim_dm, "M8", d="vb", g="vb", s="VDD", b="VDD", M=:mb, L=:L) 
add_pmos!(sim_dm, "M5", d="tail", g="vb", s="VDD", b="VDD", M=:m, L=:L)
add_pmos!(sim_dm, "M1", d="vx", g="vin", s="tail", b="VDD", M=:m, L=:L)
add_pmos!(sim_dm, "M2", d="v1", g="vip", s="tail", b="VDD", M=:m, L=:L) 
add_pmos!(sim_dm, "M7", d="vout", g="vb", s="VDD", b="VDD", M=:m_beta, L=:L)
add_nmos!(sim_dm, "M3", d="vx", g="vx", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_dm, "M4", d="v1", g="vx", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_dm, "M6", d="vout", g="v1", s="0", b="0", M=:m_2beta, L=:L)

add_res!(sim_dm, "Rz", "v1", "v_comp", Rz)
add_cap!(sim_dm, "Cc", "v_comp", "vout", Cc)
add_cap!(sim_dm, "CL", "vout", "0", CL)
compile!(sim_dm)

sim_cm = JuliaGSim(surrogate_model)
add_vsource!(sim_cm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_cm, "V_VICM", "vicm", "0", Vicm)
add_isource!(sim_cm, "I_BIAS", "vb", "0", Ib)

add_vac_source!(sim_cm, "V_VICM_AC", "vicm", "0", 1.0)

add_pmos!(sim_cm, "M8", d="vb", g="vb", s="VDD", b="VDD", M=:mb, L=:L) 
add_pmos!(sim_cm, "M5", d="tail", g="vb", s="VDD", b="VDD", M=:m, L=:L)
add_pmos!(sim_cm, "M1", d="vx", g="vicm", s="tail", b="VDD", M=:m, L=:L)
add_pmos!(sim_cm, "M2", d="v1", g="vicm", s="tail", b="VDD", M=:m, L=:L) 
add_pmos!(sim_cm, "M7", d="vout", g="vb", s="VDD", b="VDD", M=:m_beta, L=:L)
add_nmos!(sim_cm, "M3", d="vx", g="vx", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_cm, "M4", d="v1", g="vx", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_cm, "M6", d="vout", g="v1", s="0", b="0", M=:m_2beta, L=:L)

add_res!(sim_cm, "Rz", "v1", "v_comp", Rz)
add_cap!(sim_cm, "Cc", "v_comp", "vout", Cc)
add_cap!(sim_cm, "CL", "vout", "0", CL)
compile!(sim_cm)

# ==============================================================================
# 5. PARALLEL BATCH EVALUATION
# ==============================================================================
struct CircuitParams
    m::Float64; L::Float64; mb::Float64
    m_beta::Float64; m_2beta::Float64
end

M_arr = trunc.(Int, range(20, stop=50, length=20))
L_arr = range(1e-6, stop=10e-6, length=15)
combinations = vec(collect(Iterators.product(M_arr, L_arr)))
N = length(combinations) # 300

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    M_i, L_i = combinations[i]
    MB_i = trunc(Int, M_i / 2.0)
    circuits[i] = CircuitParams(Float64(M_i), Float64(L_i), Float64(MB_i), Float64(M_i * beta_val), Float64(M_i * 2.0 * beta_val))
end

freqs = 10 .^ range(0, stop=10, length=1000)

println("\n====================================================================")
println(" MASSIVE BATCH EVALUATION: $(N*2) CIRCUITS (MILLER OTA DM + CM)")
println("====================================================================")

results_x_dm = Vector{SVector{length(sim_dm.unknown_nodes), Float64}}(undef, N)
results_ac_dm = Vector{Matrix{ComplexF64}}(undef, N)
results_ac_cm = Vector{Matrix{ComplexF64}}(undef, N)

println(">>> Running $(N)-Batch DC & AC Solves on Julia Threads...")
start_time = time()

@threads for i in 1:N
    x_dm, err = solve_dc(sim_dm, circuits[i])
    results_x_dm[i] = x_dm
    results_ac_dm[i] = solve_ac(sim_dm, circuits[i], x_dm, freqs)
    
    x_cm, err_cm = solve_dc(sim_cm, circuits[i])
    results_ac_cm[i] = solve_ac(sim_cm, circuits[i], x_cm, freqs)
end

total_time = time() - start_time
println("Julia 300-Circuit Total Time: $(round(total_time, digits=4)) s")

idx_vb = findfirst(==("vb"), sim_dm.unknown_nodes)
idx_tail = findfirst(==("tail"), sim_dm.unknown_nodes)
idx_v1 = findfirst(==("v1"), sim_dm.unknown_nodes)
idx_vx = findfirst(==("vx"), sim_dm.unknown_nodes)
idx_vout_dm = findfirst(==("vout"), sim_dm.unknown_nodes)
idx_vout_cm = findfirst(==("vout"), sim_cm.unknown_nodes)

jl_vb = zeros(N); jl_tail = zeros(N); jl_v1 = zeros(N); jl_vx = zeros(N); jl_vout = zeros(N)
jl_AVD = zeros(N); jl_BW = zeros(N); jl_UGF = zeros(N); jl_PM = zeros(N); jl_AVCM = zeros(N)

for i in 1:N
    jl_vb[i] = results_x_dm[i][idx_vb]
    jl_tail[i] = results_x_dm[i][idx_tail]
    jl_v1[i] = results_x_dm[i][idx_v1]
    jl_vx[i] = results_x_dm[i][idx_vx]
    jl_vout[i] = results_x_dm[i][idx_vout_dm]

    gain_dm_db = 20 .* log10.(abs.(results_ac_dm[i][idx_vout_dm, :]))
    gain_cm_db = 20 .* log10.(abs.(results_ac_cm[i][idx_vout_cm, :]))
    phase_dm = unwrap_phase_custom(angle.(results_ac_dm[i][idx_vout_dm, :])) .* 180 ./ pi
    
    jl_AVD[i] = maximum(gain_dm_db)
    jl_AVCM[i] = gain_cm_db[1]
    
    bw_idx = findfirst(x -> x <= jl_AVD[i] - 3.0, gain_dm_db)
    jl_BW[i] = bw_idx !== nothing ? freqs[bw_idx] : freqs[end]
    
    ugf_i = argmin(abs.(gain_dm_db))
    jl_UGF[i] = freqs[ugf_i]
    jl_PM[i] = 180.0 + phase_dm[ugf_i]
end

# ==============================================================================
# 6. LTSPICE VERIFICATION & PLOTTING
# ==============================================================================
println("\n>>> Running SPICE (Sequential) for all 300 points (DC & AC)...")
BASE_DIR = pwd()

spice_vb = zeros(N); spice_tail = zeros(N); spice_v1 = zeros(N); spice_vx = zeros(N); spice_vout = zeros(N)
spice_AVD = zeros(N); spice_BW = zeros(N); spice_UGF = zeros(N)
spice_PM = zeros(N); spice_AVCM = zeros(N); spice_CMRR = zeros(N)
spice_traces = Dict{Int, Dict}()

start_spice = time()
for i in 1:N
    if i % 10 == 0 || i == N
        print("\r SPICE Progress: $i/$N... ($(round((i/N)*100, digits=1))%)")
        flush(stdout)
    end
    
    M_i = trunc(Int, circuits[i].m)
    L_i = circuits[i].L
    MB_i = trunc(Int, circuits[i].mb)
    
    try
        v_dc = evaluate_miller_ota_dc(M_i, L_i, M_i, L_i, M_i, L_i, MB_i, beta_val, Ib, Vicm, BASE_DIR)
        spice_vb[i] = pyconvert(Float64, v_dc["vb"])
        spice_tail[i] = pyconvert(Float64, v_dc["tail"])
        spice_v1[i] = pyconvert(Float64, v_dc["v1"])
        spice_vx[i] = pyconvert(Float64, v_dc["vx"])
        spice_vout[i] = pyconvert(Float64, v_dc["vout"])
    
        ac_specs = evaluate_miller_ota_ac(M_i, L_i, M_i, L_i, M_i, L_i, MB_i, beta_val, Ib, Cc, Rz, CL, Vicm, BASE_DIR)
        spice_AVD[i], spice_BW[i] = pyconvert(Float64, ac_specs["AVD"]), pyconvert(Float64, ac_specs["BW"])
        spice_UGF[i], spice_PM[i] = pyconvert(Float64, ac_specs["UGF"]), pyconvert(Float64, ac_specs["PM"])
        spice_AVCM[i], spice_CMRR[i] = pyconvert(Float64, ac_specs["AVCM"]), pyconvert(Float64, ac_specs["CMRR"])
        
        spice_traces[i] = Dict(
            "freqs" => pyconvert(Array, ac_specs["freqs"]),
            "mag"   => pyconvert(Array, ac_specs["mag"]),
            "phase" => pyconvert(Array, ac_specs["phase"])
        )
    catch e
        if i == 1
            @error "LTspice failed on first iteration: " exception=e
        end
    end
end
println("\nSPICE evaluation complete!")
spice_time = time() - start_spice
println("SPICE Total Time: $(round(spice_time, digits=4)) s")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")

err_vb = abs.(jl_vb .- spice_vb)
err_tail = abs.(jl_tail .- spice_tail)
err_v1 = abs.(jl_v1 .- spice_v1)
err_vx = abs.(jl_vx .- spice_vx)
err_vout = abs.(jl_vout .- spice_vout)

max_err_circuit = max.(err_vb, err_tail, err_v1, err_vx, err_vout)
worst_idx = sortperm(max_err_circuit, rev=true)[1:10]

plt = pyimport("matplotlib.pyplot")

println("\n======================================================================")
println(" TOP 10 WORST CASE CIRCUITS: JULIA VS LTSPICE (DC & AC)")
println("======================================================================")
for (rank, idx) in enumerate(worst_idx)
    M_i = trunc(Int, circuits[idx].m)
    L_i = circuits[idx].L
    MB_i = trunc(Int, circuits[idx].mb)
    
    @printf("Rank %d (Circuit #%d) - Max DC Error: %.3f mV\n", rank, idx-1, max_err_circuit[idx]*1000)
    @printf("Params: M=%d (W=%.2fum), L=%.2fum, MB=%d (WB=%.2fum)\n\n", M_i, M_i*0.5, L_i*1e6, MB_i, MB_i*0.5)
    
    println("  DC Nodes | Julia (V)   | LTspice (V) | Diff (mV)")
    @printf("  vb       | %.4f      | %.4f      | %.3f\n", jl_vb[idx], spice_vb[idx], err_vb[idx]*1000)
    @printf("  tail     | %.4f      | %.4f      | %.3f\n", jl_tail[idx], spice_tail[idx], err_tail[idx]*1000)
    @printf("  v1       | %.4f      | %.4f      | %.3f\n", jl_v1[idx], spice_v1[idx], err_v1[idx]*1000)
    @printf("  vx       | %.4f      | %.4f      | %.3f\n", jl_vx[idx], spice_vx[idx], err_vx[idx]*1000)
    @printf("  vout     | %.4f      | %.4f      | %.3f\n\n", jl_vout[idx], spice_vout[idx], err_vout[idx]*1000)
    
    println("  AC Specs | Julia       | LTspice     | Diff")
    @printf("  AVD      | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx], spice_AVD[idx], abs(jl_AVD[idx] - spice_AVD[idx]))
    diff_bw = abs(jl_BW[idx] - spice_BW[idx]) / (spice_BW[idx] + 1e-12) * 100
    @printf("  BW       | %.2e Hz | %.2e Hz | %.2f %%\n", jl_BW[idx], spice_BW[idx], diff_bw)
    diff_ugf = abs(jl_UGF[idx] - spice_UGF[idx]) / (spice_UGF[idx] + 1e-12) * 100
    @printf("  UGF      | %.2e Hz | %.2e Hz | %.2f %%\n", jl_UGF[idx], spice_UGF[idx], diff_ugf)
    @printf("  PM       | %6.2f deg  | %6.2f deg  | %.2f deg\n", jl_PM[idx], spice_PM[idx], abs(jl_PM[idx] - spice_PM[idx]))
    @printf("  AVCM     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVCM[idx], spice_AVCM[idx], abs(jl_AVCM[idx] - spice_AVCM[idx]))
    @printf("  CMRR     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx]-jl_AVCM[idx], spice_CMRR[idx], abs((jl_AVD[idx]-jl_AVCM[idx]) - spice_CMRR[idx]))
    
    gain_dm_db_idx = 20 .* log10.(abs.(results_ac_dm[idx][idx_vout_dm, :]))
    phase_dm_idx = unwrap_phase_custom(angle.(results_ac_dm[idx][idx_vout_dm, :])) .* 180 ./ pi
    freqs_array = collect(freqs) 
    
    plt.figure(figsize=(10, 8))
    
    plt.subplot(2, 1, 1)
    plt.semilogx(freqs_array, gain_dm_db_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["mag"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.title("Rank $rank (Circuit #$(idx-1)) - AC Response")
    plt.ylabel("Magnitude (dB)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.subplot(2, 1, 2)
    plt.semilogx(freqs_array, phase_dm_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["phase"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.ylabel("Phase (deg)")
    plt.xlabel("Frequency (Hz)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.tight_layout()
    plt.savefig("temp_plot_miller.png", bbox_inches="tight")
    display(MIME("image/png"), read("temp_plot_miller.png"))
    plt.close()
    
    println("-" ^ 70)
end
```
