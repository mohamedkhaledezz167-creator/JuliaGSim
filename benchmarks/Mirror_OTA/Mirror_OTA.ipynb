```julia
# ==============================================================================
# 1. Call the LUT definitions and the simulator Engine
# ==============================================================================
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"

using Random
using Printf
using StaticArrays
using PythonCall

# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim.jl")

# ==============================================================================
# 2.Helper to unwrap phase angles natively in Julia 
# ==============================================================================
function unwrap_phase_custom(p)
    p_out = copy(p)
    for j in 2:length(p_out)
        d = p_out[j] - p_out[j-1]
        p_out[j] -= 2pi * round(d / 2pi) 
    end
    return p_out
end

# ==============================================================================
# 3. LTSPICE VERIFICATION GROUND TRUTH (Self-Contained Dynamic Netlists)
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, subprocess, numpy as np
from PyLTSpice import RawRead

def evaluate_mirror_ota_dc_voltages(md, Ld, mm, Lm, ml, Ll, mss, Lss, mb, Ib, alpha, beta, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR, "Mirror_OTA_N_DC.sp")
    rp = os.path.join(BASE_DIR, "Mirror_OTA_N_DC.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')

    mm_m5 = alpha * mm
    mm_m6 = beta * mm
    ml_m7 = (alpha / beta) * ml

    nl = f'''* Mirror OTA (NMOS Input) - DC Operating Point AutoGen
Vdd Vdd 0 3
Vicm vcm 0 {Vicm}
vinp vip vcm 0
vinn vin vcm 0

IB Vdd vb_n {Ib}
M10 vb_n vb_n 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}
M9 vT vb_n 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}

M1 v_diode1 vin vT 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2 v_diode2 vip vT 0 nmos_3p3 l={Ld} w=0.5u m={md}

M3 v_diode1 v_diode1 Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M4 v_diode2 v_diode2 Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M5 v_mir1 v_diode1 Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m5}
M6 vout v_diode2 Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m6}

M7 v_mir1 v_mir1 0 0 nmos_3p3 l={Ll} w=0.5u m={ml_m7}
M8 vout v_mir1 0 0 nmos_3p3 l={Ll} w=0.5u m={ml}

CL vout 0 1e-12

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.op
.end
'''
    with open(sp, 'w', encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp], cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(rp)
    tr = raw.get_trace_names()
    def gv(n):
        for t in tr:
            if t.lower() == n.lower(): return float(raw.get_trace(t).get_wave(0)[0])
        return 0.0
    return {k: gv(f"V({k})") for k in ['vb_n', 'vT', 'v_diode1', 'v_diode2', 'v_mir1', 'vout']}

def evaluate_mirror_ota_ac_ltspice(md, Ld, mm, Lm, ml, Ll, mss, Lss, mb, Ib, alpha, beta, CL, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR, "Mirror_OTA_AC_TEST.sp")
    rp = os.path.join(BASE_DIR, "Mirror_OTA_AC_TEST.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')
    
    mm_m5 = alpha * mm
    mm_m6 = beta * mm
    ml_m7 = (alpha / beta) * ml
    
    nl = f'''* Mirror OTA AC AutoGen
Vdd Vdd 0 3
IB_dm Vdd vb_n_dm {Ib}
vicm_dm vcm_dm 0 {Vicm}
vinp_dm vip_dm vcm_dm AC 0.5
vinn_dm vin_dm vcm_dm AC -0.5

M10_dm vb_n_dm vb_n_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb} ad=0 as=0 pd=0 ps=0
M9_dm vT_dm vb_n_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss} ad=0 as=0 pd=0 ps=0
M1_dm v_diode1_dm vin_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M2_dm v_diode2_dm vip_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M3_dm v_diode1_dm v_diode1_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M4_dm v_diode2_dm v_diode2_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M5_dm v_mir1_dm v_diode1_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m5} ad=0 as=0 pd=0 ps=0
M6_dm vout_dm v_diode2_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m6} ad=0 as=0 pd=0 ps=0
M7_dm v_mir1_dm v_mir1_dm 0 0 nmos_3p3 l={Ll} w=0.5u m={ml_m7} ad=0 as=0 pd=0 ps=0
M8_dm vout_dm v_mir1_dm 0 0 nmos_3p3 l={Ll} w=0.5u m={ml} ad=0 as=0 pd=0 ps=0
CL_dm vout_dm 0 {CL}

IB_cm Vdd vb_n_cm {Ib}
vicm_ac vicm_ac 0 {Vicm} AC 1

M10_cm vb_n_cm vb_n_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb} ad=0 as=0 pd=0 ps=0
M9_cm vT_cm vb_n_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss} ad=0 as=0 pd=0 ps=0
M1_cm v_diode1_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M2_cm v_diode2_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md} ad=0 as=0 pd=0 ps=0
M3_cm v_diode1_cm v_diode1_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M4_cm v_diode2_cm v_diode2_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm} ad=0 as=0 pd=0 ps=0
M5_cm v_mir1_cm v_diode1_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m5} ad=0 as=0 pd=0 ps=0
M6_cm vout_cm v_diode2_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm_m6} ad=0 as=0 pd=0 ps=0
M7_cm v_mir1_cm v_mir1_cm 0 0 nmos_3p3 l={Ll} w=0.5u m={ml_m7} ad=0 as=0 pd=0 ps=0
M8_cm vout_cm v_mir1_cm 0 0 nmos_3p3 l={Ll} w=0.5u m={ml} ad=0 as=0 pd=0 ps=0
CL_cm vout_cm 0 {CL}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.ac dec 100 1 1e9
.end
'''
    with open(sp, 'w', encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH, "-b", "-run", sp], cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(rp)
    freqs = np.abs(raw.get_trace("frequency").get_wave(0))
    
    vout_dm_c = raw.get_trace("V(vout_dm)").get_wave(0)
    vout_cm_c = raw.get_trace("V(vout_cm)").get_wave(0)
    
    dm = 20 * np.log10(np.abs(vout_dm_c))
    cm = 20 * np.log10(np.abs(vout_cm_c))
    phase = np.unwrap(np.angle(vout_dm_c)) * 180 / np.pi
    
    AVD = np.max(dm)
    bwi = np.where(dm <= AVD - 3.0)[0]
    BW = freqs[bwi[0]] if len(bwi) else freqs[-1]
    ugf_i = np.argmin(np.abs(dm))
    UGF = freqs[ugf_i]
    PM = 180.0 + phase[ugf_i]
    AVCM = cm[0]
    
    return {'AVD': AVD, 'BW': BW, 'UGF': UGF, 'PM': PM, 'AVCM': AVCM, 'CMRR': AVD - AVCM,
            'freqs': freqs, 'mag': dm, 'phase': phase}
""", g)
evaluate_mirror_ota_dc = g["evaluate_mirror_ota_dc_voltages"]
evaluate_mirror_ota_ac = g["evaluate_mirror_ota_ac_ltspice"]

# ==============================================================================
# 4. BUILDING THE DIFFERENTIAL AND COMMON MODE MIRROR OTA
# ==============================================================================
VDD, Vicm, Ib = 3.0, 1.5, 2e-5
alpha, beta = 1.0, 2.0
CL = 1e-12

sim_dm = JuliaGSim(surrogate_model)
add_vsource!(sim_dm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_dm, "V_VIP", "vip", "0", Vicm)
add_vsource!(sim_dm, "V_VIN", "vin", "0", Vicm)
add_isource!(sim_dm, "I_BIAS", "VDD", "vb_n", Ib)

add_vac_source!(sim_dm, "V_VIP_AC", "vip", "0", 0.5)
add_vac_source!(sim_dm, "V_VIN_AC", "vin", "0", -0.5)

add_nmos!(sim_dm, "M10", d="vb_n", g="vb_n", s="0", b="0", M=:mb, L=:L)
add_nmos!(sim_dm, "M9",  d="vT", g="vb_n", s="0", b="0", M=:mss, L=:L)
add_nmos!(sim_dm, "M1",  d="v_diode1", g="vin", s="vT", b="0", M=:md, L=:L)
add_nmos!(sim_dm, "M2",  d="v_diode2", g="vip", s="vT", b="0", M=:md, L=:L)
add_pmos!(sim_dm, "M3",  d="v_diode1", g="v_diode1", s="VDD", b="VDD", M=:mm, L=:L)
add_pmos!(sim_dm, "M4",  d="v_diode2", g="v_diode2", s="VDD", b="VDD", M=:mm, L=:L)
add_pmos!(sim_dm, "M5",  d="v_mir1",   g="v_diode1", s="VDD", b="VDD", M=:mm_m5, L=:L)
add_pmos!(sim_dm, "M6",  d="vout",     g="v_diode2", s="VDD", b="VDD", M=:mm_m6, L=:L)
add_nmos!(sim_dm, "M7",  d="v_mir1", g="v_mir1", s="0", b="0", M=:ml_m7, L=:L)
add_nmos!(sim_dm, "M8",  d="vout",   g="v_mir1", s="0", b="0", M=:ml, L=:L)
add_cap!(sim_dm, "CL", "vout", "0", CL)
compile!(sim_dm)

sim_cm = JuliaGSim(surrogate_model)
add_vsource!(sim_cm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_cm, "V_VICM", "vicm", "0", Vicm)
add_isource!(sim_cm, "I_BIAS", "VDD", "vb_n", Ib)

add_vac_source!(sim_cm, "V_VICM_AC", "vicm", "0", 1.0)

add_nmos!(sim_cm, "M10", d="vb_n", g="vb_n", s="0", b="0", M=:mb, L=:L)
add_nmos!(sim_cm, "M9",  d="vT", g="vb_n", s="0", b="0", M=:mss, L=:L)
add_nmos!(sim_cm, "M1",  d="v_diode1", g="vicm", s="vT", b="0", M=:md, L=:L)
add_nmos!(sim_cm, "M2",  d="v_diode2", g="vicm", s="vT", b="0", M=:md, L=:L)
add_pmos!(sim_cm, "M3",  d="v_diode1", g="v_diode1", s="VDD", b="VDD", M=:mm, L=:L)
add_pmos!(sim_cm, "M4",  d="v_diode2", g="v_diode2", s="VDD", b="VDD", M=:mm, L=:L)
add_pmos!(sim_cm, "M5",  d="v_mir1",   g="v_diode1", s="VDD", b="VDD", M=:mm_m5, L=:L)
add_pmos!(sim_cm, "M6",  d="vout",     g="v_diode2", s="VDD", b="VDD", M=:mm_m6, L=:L)
add_nmos!(sim_cm, "M7",  d="v_mir1", g="v_mir1", s="0", b="0", M=:ml_m7, L=:L)
add_nmos!(sim_cm, "M8",  d="vout",   g="v_mir1", s="0", b="0", M=:ml, L=:L)
add_cap!(sim_cm, "CL", "vout", "0", CL)
compile!(sim_cm)

# ==============================================================================
# 5. PARALLEL BATCH EVALUATION
# ==============================================================================
struct CircuitParams
    md::Float64; mss::Float64; mm::Float64
    ml::Float64; mb::Float64
    mm_m5::Float64; mm_m6::Float64; ml_m7::Float64
    L::Float64
end

W_arr = range(20e-6, stop=40e-6, length=20)
L_arr = range(1e-6, stop=10e-6, length=15)
combinations = vec(collect(Iterators.product(L_arr, W_arr)))
N = length(combinations) # 300

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    L_i, W_i = combinations[i]
    
    md_i = W_i / 0.5e-6
    mss_i = W_i / 0.5e-6
    mm_i = 0.5 * (W_i / 0.5e-6)
    ml_i = 0.5 * (W_i / 0.5e-6)
    mb_i = 0.5 * (W_i / 0.5e-6)
    
    mm_m5_i = alpha * mm_i
    mm_m6_i = beta * mm_i
    ml_m7_i = (alpha / beta) * ml_i
    
    circuits[i] = CircuitParams(md_i, mss_i, mm_i, ml_i, mb_i, mm_m5_i, mm_m6_i, ml_m7_i, L_i)
end

freqs = 10 .^ range(0, stop=9, length=900)

println("\n====================================================================")
println(" MASSIVE BATCH EVALUATION: $(N*2) CIRCUITS (MIRROR OTA DM + CM)")
println("====================================================================")

results_x_dm = Vector{SVector{length(sim_dm.unknown_nodes), Float64}}(undef, N)
results_ac_dm = Vector{Matrix{ComplexF64}}(undef, N)
results_ac_cm = Vector{Matrix{ComplexF64}}(undef, N)

println(">>> Running $(N)-Batch DC & AC Solves on Julia Threads...")
start_time = time()

@threads for i in 1:N
    x_dm, err = solve_dc(sim_dm, circuits[i])
    results_x_dm[i] = x_dm
    results_ac_dm[i] = solve_ac(sim_dm, circuits[i], x_dm, freqs)
    
    x_cm, err_cm = solve_dc(sim_cm, circuits[i])
    results_ac_cm[i] = solve_ac(sim_cm, circuits[i], x_cm, freqs)
end

total_time = time() - start_time
println("Julia 300-Circuit Total Time: $(round(total_time, digits=4)) s")

idx_vb_n = findfirst(==("vb_n"), sim_dm.unknown_nodes)
idx_vT = findfirst(==("vT"), sim_dm.unknown_nodes)
idx_vd1 = findfirst(==("v_diode1"), sim_dm.unknown_nodes)
idx_vd2 = findfirst(==("v_diode2"), sim_dm.unknown_nodes)
idx_vm1 = findfirst(==("v_mir1"), sim_dm.unknown_nodes)
idx_vout_dm = findfirst(==("vout"), sim_dm.unknown_nodes)
idx_vout_cm = findfirst(==("vout"), sim_cm.unknown_nodes)

jl_vb_n = zeros(N); jl_vT = zeros(N); jl_vd1 = zeros(N); jl_vd2 = zeros(N); jl_vm1 = zeros(N); jl_vout = zeros(N)
jl_AVD = zeros(N); jl_BW = zeros(N); jl_UGF = zeros(N); jl_PM = zeros(N); jl_AVCM = zeros(N)

for i in 1:N
    jl_vb_n[i] = results_x_dm[i][idx_vb_n]
    jl_vT[i] = results_x_dm[i][idx_vT]
    jl_vd1[i] = results_x_dm[i][idx_vd1]
    jl_vd2[i] = results_x_dm[i][idx_vd2]
    jl_vm1[i] = results_x_dm[i][idx_vm1]
    jl_vout[i] = results_x_dm[i][idx_vout_dm]

    gain_dm_db = 20 .* log10.(abs.(results_ac_dm[i][idx_vout_dm, :]))
    gain_cm_db = 20 .* log10.(abs.(results_ac_cm[i][idx_vout_cm, :]))
    phase_dm = unwrap_phase_custom(angle.(results_ac_dm[i][idx_vout_dm, :])) .* 180 ./ pi
    
    jl_AVD[i] = maximum(gain_dm_db)
    jl_AVCM[i] = gain_cm_db[1]
    
    bw_idx = findfirst(x -> x <= jl_AVD[i] - 3.0, gain_dm_db)
    jl_BW[i] = bw_idx !== nothing ? freqs[bw_idx] : freqs[end]
    
    ugf_i = argmin(abs.(gain_dm_db))
    jl_UGF[i] = freqs[ugf_i]
    jl_PM[i] = 180.0 + phase_dm[ugf_i]
end

# ==============================================================================
# 6. LTSPICE VERIFICATION & PLOTTING
# ==============================================================================
println("\n>>> Running SPICE (Sequential) for all 300 points (DC & AC)...")
BASE_DIR = pwd()

spice_vb_n = zeros(N); spice_vT = zeros(N); spice_vd1 = zeros(N); spice_vd2 = zeros(N); spice_vm1 = zeros(N); spice_vout = zeros(N)
spice_AVD = zeros(N); spice_BW = zeros(N); spice_UGF = zeros(N)
spice_PM = zeros(N); spice_AVCM = zeros(N); spice_CMRR = zeros(N)
spice_traces = Dict{Int, Dict}()

start_spice = time()
for i in 1:N
    if i % 10 == 0 || i == N
        print("\r SPICE Progress: $i/$N... ($(round((i/N)*100, digits=1))%)")
        flush(stdout)
    end
    
    md_i = circuits[i].md
    mss_i = circuits[i].mss
    mm_i = circuits[i].mm
    ml_i = circuits[i].ml
    mb_i = circuits[i].mb
    L_i = circuits[i].L
    
    try
        v_dc = evaluate_mirror_ota_dc(md_i, L_i, mm_i, L_i, ml_i, L_i, mss_i, L_i, mb_i, Ib, alpha, beta, Vicm, BASE_DIR)
        spice_vb_n[i] = pyconvert(Float64, v_dc["vb_n"])
        spice_vT[i] = pyconvert(Float64, v_dc["vT"])
        spice_vd1[i] = pyconvert(Float64, v_dc["v_diode1"])
        spice_vd2[i] = pyconvert(Float64, v_dc["v_diode2"])
        spice_vm1[i] = pyconvert(Float64, v_dc["v_mir1"])
        spice_vout[i] = pyconvert(Float64, v_dc["vout"])
    
        ac_specs = evaluate_mirror_ota_ac(md_i, L_i, mm_i, L_i, ml_i, L_i, mss_i, L_i, mb_i, Ib, alpha, beta, CL, Vicm, BASE_DIR)
        spice_AVD[i], spice_BW[i] = pyconvert(Float64, ac_specs["AVD"]), pyconvert(Float64, ac_specs["BW"])
        spice_UGF[i], spice_PM[i] = pyconvert(Float64, ac_specs["UGF"]), pyconvert(Float64, ac_specs["PM"])
        spice_AVCM[i], spice_CMRR[i] = pyconvert(Float64, ac_specs["AVCM"]), pyconvert(Float64, ac_specs["CMRR"])
        
        spice_traces[i] = Dict(
            "freqs" => pyconvert(Array, ac_specs["freqs"]),
            "mag"   => pyconvert(Array, ac_specs["mag"]),
            "phase" => pyconvert(Array, ac_specs["phase"])
        )
    catch e
        if i == 1
            @error "LTspice failed on first iteration: " exception=e
        end
    end
end
println("\nSPICE evaluation complete!")
spice_time = time() - start_spice
println("SPICE Total Time: $(round(spice_time, digits=4)) s")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")

err_vb_n = abs.(jl_vb_n .- spice_vb_n)
err_vT = abs.(jl_vT .- spice_vT)
err_vd1 = abs.(jl_vd1 .- spice_vd1)
err_vd2 = abs.(jl_vd2 .- spice_vd2)
err_vm1 = abs.(jl_vm1 .- spice_vm1)
err_vout = abs.(jl_vout .- spice_vout)

max_err_circuit = max.(err_vb_n, err_vT, err_vd1, err_vd2, err_vm1, err_vout)
worst_idx = sortperm(max_err_circuit, rev=true)[1:10]

plt = pyimport("matplotlib.pyplot")

println("\n======================================================================")
println(" TOP 10 WORST CASE CIRCUITS: JULIA VS LTSPICE (DC & AC)")
println("======================================================================")
for (rank, idx) in enumerate(worst_idx)
    md_i = circuits[idx].md
    L_i = circuits[idx].L
    
    W_i = md_i * 0.5e-6
    
    @printf("Rank %d (Circuit #%d) - Max DC Error: %.3f mV\n", rank, idx-1, max_err_circuit[idx]*1000)
    @printf("Params: W=%.2fum, L=%.2fum\n\n", W_i*1e6, L_i*1e6)
    
    println("  DC Nodes | Julia (V)   | LTspice (V) | Diff (mV)")
    @printf("  vb_n     | %.4f      | %.4f      | %.3f\n", jl_vb_n[idx], spice_vb_n[idx], err_vb_n[idx]*1000)
    @printf("  vT       | %.4f      | %.4f      | %.3f\n", jl_vT[idx], spice_vT[idx], err_vT[idx]*1000)
    @printf("  v_diode1 | %.4f      | %.4f      | %.3f\n", jl_vd1[idx], spice_vd1[idx], err_vd1[idx]*1000)
    @printf("  v_diode2 | %.4f      | %.4f      | %.3f\n", jl_vd2[idx], spice_vd2[idx], err_vd2[idx]*1000)
    @printf("  v_mir1   | %.4f      | %.4f      | %.3f\n", jl_vm1[idx], spice_vm1[idx], err_vm1[idx]*1000)
    @printf("  vout     | %.4f      | %.4f      | %.3f\n\n", jl_vout[idx], spice_vout[idx], err_vout[idx]*1000)
    
    println("  AC Specs | Julia       | LTspice     | Diff")
    @printf("  AVD      | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx], spice_AVD[idx], abs(jl_AVD[idx] - spice_AVD[idx]))
    diff_bw = abs(jl_BW[idx] - spice_BW[idx]) / (spice_BW[idx] + 1e-12) * 100
    @printf("  BW       | %.2e Hz | %.2e Hz | %.2f %%\n", jl_BW[idx], spice_BW[idx], diff_bw)
    diff_ugf = abs(jl_UGF[idx] - spice_UGF[idx]) / (spice_UGF[idx] + 1e-12) * 100
    @printf("  UGF      | %.2e Hz | %.2e Hz | %.2f %%\n", jl_UGF[idx], spice_UGF[idx], diff_ugf)
    @printf("  PM       | %6.2f deg  | %6.2f deg  | %.2f deg\n", jl_PM[idx], spice_PM[idx], abs(jl_PM[idx] - spice_PM[idx]))
    @printf("  AVCM     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVCM[idx], spice_AVCM[idx], abs(jl_AVCM[idx] - spice_AVCM[idx]))
    @printf("  CMRR     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx]-jl_AVCM[idx], spice_CMRR[idx], abs((jl_AVD[idx]-jl_AVCM[idx]) - spice_CMRR[idx]))
    
    gain_dm_db_idx = 20 .* log10.(abs.(results_ac_dm[idx][idx_vout_dm, :]))
    phase_dm_idx = unwrap_phase_custom(angle.(results_ac_dm[idx][idx_vout_dm, :])) .* 180 ./ pi
    freqs_array = collect(freqs) 
    
    plt.figure(figsize=(10, 8))
    
    plt.subplot(2, 1, 1)
    plt.semilogx(freqs_array, gain_dm_db_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["mag"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.title("Rank $rank (Circuit #$(idx-1)) - AC Response")
    plt.ylabel("Magnitude (dB)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.subplot(2, 1, 2)
    plt.semilogx(freqs_array, phase_dm_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["phase"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.ylabel("Phase (deg)")
    plt.xlabel("Frequency (Hz)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.tight_layout()
    plt.savefig("temp_plot_mirror.png", bbox_inches="tight")
    display(MIME("image/png"), read("temp_plot_mirror.png"))
    plt.close()
    
    println("-" ^ 70)
end
```
