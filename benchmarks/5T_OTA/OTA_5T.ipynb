```julia
# ==============================================================================
# 1. Call the LUT definitions and the simulator Engine
# ==============================================================================
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"

using Random
using Printf
using StaticArrays
using PythonCall

# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim_Finalized.jl")

# ==============================================================================
# 2.Helper to unwrap phase angles natively in Julia 
# ==============================================================================
function unwrap_phase_custom(p)
    p_out = copy(p)
    for j in 2:length(p_out)
        d = p_out[j] - p_out[j-1]
        p_out[j] -= 2pi * round(d / 2pi) 
    end
    return p_out
end

# ==============================================================================
# 3. LTSPICE VERIFICATION GROUND TRUTH (Self-Contained Dynamic Netlists)
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, subprocess, numpy as np
from PyLTSpice import RawRead

def evaluate_5t_ota_dc_voltages(md, Ld, mm, Lm, mss, Lss, mb, Ib, CL, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    netlist_path = os.path.join(BASE_DIR, "OTA_5T_N_DC.sp")
    raw_path = os.path.join(BASE_DIR, "OTA_5T_N_DC.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')

    netlist = f'''* 5T-OTA - DC Operating Point AutoGen
Vdd Vdd 0 3
Vicm vcm 0 {Vicm}
vinp vip vcm 0
vinn vin vcm 0

IB Vdd vb {Ib}
CL vout 0 {CL}

M1 vm vip vT 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2 vout vin vT 0 nmos_3p3 l={Ld} w=0.5u m={md}

M3 vm vm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M4 vout vm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}

M5 vT vb 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M6 vb vb 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.op
.end
'''
    with open(netlist_path, 'w', encoding='utf-8') as f:
        f.write(netlist)
        
    cmd = [LTSPICE_PATH, "-b", "-run", netlist_path]
    subprocess.run(cmd, cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(raw_path)
    traces = raw.get_trace_names()
    def get_v(name):
        for t in traces:
            if t.lower() == name.lower():
                return float(raw.get_trace(t).get_wave(0)[0])
        return 0.0
    return {'vb': get_v("V(vb)"), 'vm': get_v("V(vm)"), 'vT': get_v("V(vT)"), 'vout': get_v("V(vout)")}

def evaluate_5t_ota_ac_ltspice(md, Ld, mm, Lm, mss, Lss, mb, Ib, CL, Vicm, BASE_DIR='.'):
    os.makedirs(BASE_DIR, exist_ok=True)
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    netlist_path = os.path.join(BASE_DIR, "OTA_5T_AC_TEST.sp")
    raw_path = os.path.join(BASE_DIR, "OTA_5T_AC_TEST.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA', ''), 'LTspice', 'lib', 'cmp', 'standard.mos')

    netlist = f'''* 5T-OTA - AC Testbench AutoGen
Vdd Vdd 0 3
IB_dm Vdd vb_dm {Ib}
vicm_dm vcm_dm 0 {Vicm}
vinp_dm vip_dm vcm_dm AC 0.5
vinn_dm vin_dm vcm_dm AC -0.5
M1_dm vm_dm vip_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2_dm vout_dm vin_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M3_dm vm_dm vm_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M4_dm vout_dm vm_dm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M5_dm vT_dm vb_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M6_dm vb_dm vb_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}
CL_dm vout_dm 0 {CL}

IB_cm Vdd vb_cm {Ib}
vicm_ac vicm_ac 0 {Vicm} AC 1
M1_cm vm_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2_cm vout_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M3_cm vm_cm vm_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M4_cm vout_cm vm_cm Vdd Vdd pmos_3p3 l={Lm} w=0.5u m={mm}
M5_cm vT_cm vb_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M6_cm vb_cm vb_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}
CL_cm vout_cm 0 {CL}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt

.ac dec 100 1 1e9
.end
'''
    with open(netlist_path, 'w', encoding='utf-8') as f: f.write(netlist)    
    cmd = [LTSPICE_PATH, "-b", "-run", netlist_path]
    subprocess.run(cmd, cwd=BASE_DIR, capture_output=True, text=True)
    raw = RawRead(raw_path)
    freqs = np.abs(raw.get_trace("frequency").get_wave(0))
    vout_dm_c = raw.get_trace("V(vout_dm)").get_wave(0)
    vout_dm = np.abs(vout_dm_c)
    vout_cm = np.abs(raw.get_trace("V(vout_cm)").get_wave(0))
    
    gain_dm_db = 20 * np.log10(vout_dm)
    gain_cm_db = 20 * np.log10(vout_cm)
    phase = np.unwrap(np.angle(vout_dm_c)) * 180 / np.pi
    
    AVD = np.max(gain_dm_db)
    bw_idx = np.where(gain_dm_db <= AVD - 3.0)[0]
    BW = freqs[bw_idx[0]] if len(bw_idx) > 0 else freqs[-1]
    ugf_i = np.argmin(np.abs(gain_dm_db))
    UGF = freqs[ugf_i]
    PM = 180.0 + phase[ugf_i]
    AVCM = gain_cm_db[0]
    
    return {'AVD': AVD, 'BW': BW, 'UGF': UGF, 'PM': PM, 'AVCM': AVCM, 'CMRR': AVD - AVCM,
            'freqs': freqs, 'mag': gain_dm_db, 'phase': phase}
""", g)
evaluate_5t_ota_dc = g["evaluate_5t_ota_dc_voltages"]
evaluate_5t_ota_ac = g["evaluate_5t_ota_ac_ltspice"]

# ==============================================================================
# 4. BUILDING THE DIFFERENTIAL AND COMMON MODE 5T-OTA
# ==============================================================================
VDD, Vicm, Ib, CL = 3.0, 1.8, 1e-5, 0.1e-12

sim_dm = JuliaGSim(surrogate_model)
add_vsource!(sim_dm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_dm, "V_VIP", "vip", "0", Vicm)
add_vsource!(sim_dm, "V_VIN", "vin", "0", Vicm)
add_isource!(sim_dm, "I_BIAS", "VDD", "vb", Ib)

add_vac_source!(sim_dm, "V_VIP_AC", "vip", "0", 0.5)
add_vac_source!(sim_dm, "V_VIN_AC", "vin", "0", -0.5)

add_nmos!(sim_dm, "M1", d="vm", g="vip", s="vT", b="0", M=:m, L=:L)
add_nmos!(sim_dm, "M2", d="vout", g="vin", s="vT", b="0", M=:m, L=:L)
add_pmos!(sim_dm, "M3", d="vm", g="vm", s="VDD", b="VDD", M=:m, L=:L)
add_pmos!(sim_dm, "M4", d="vout", g="vm", s="VDD", b="VDD", M=:m, L=:L)
add_nmos!(sim_dm, "M5", d="vT", g="vb", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_dm, "M6", d="vb", g="vb", s="0", b="0", M=:mb, L=:L)
add_cap!(sim_dm, "CL", "vout", "0", CL)
compile!(sim_dm)

sim_cm = JuliaGSim(surrogate_model)
add_vsource!(sim_cm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_cm, "V_VICM", "vicm", "0", Vicm)
add_isource!(sim_cm, "I_BIAS", "VDD", "vb", Ib)

add_vac_source!(sim_cm, "V_VICM_AC", "vicm", "0", 1.0)

add_nmos!(sim_cm, "M1", d="vm", g="vicm", s="vT", b="0", M=:m, L=:L)
add_nmos!(sim_cm, "M2", d="vout", g="vicm", s="vT", b="0", M=:m, L=:L)
add_pmos!(sim_cm, "M3", d="vm", g="vm", s="VDD", b="VDD", M=:m, L=:L)
add_pmos!(sim_cm, "M4", d="vout", g="vm", s="VDD", b="VDD", M=:m, L=:L)
add_nmos!(sim_cm, "M5", d="vT", g="vb", s="0", b="0", M=:m, L=:L)
add_nmos!(sim_cm, "M6", d="vb", g="vb", s="0", b="0", M=:mb, L=:L)
add_cap!(sim_cm, "CL", "vout", "0", CL)
compile!(sim_cm)

# ==============================================================================
# 5. PARALLEL BATCH EVALUATION
# ==============================================================================
struct CircuitParams
    m::Float64; L::Float64; mb::Float64
end

# FIXED: use trunc.(Int, ...) to EXACTLY match Python's .astype(int) behavior
M_arr = trunc.(Int, range(20, 60, length=10))
L_arr = range(0.5e-6, 5e-6, length=10)
MB_arr = trunc.(Int, range(1, 30, length=10))

combinations = vec(collect(Iterators.product(M_arr, L_arr, MB_arr)))
Random.seed!(42)
sampled_combinations = combinations[randperm(length(combinations))[1:300]]
N = 300

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    M_i, L_i, MB_i = sampled_combinations[i]
    circuits[i] = CircuitParams(M_i, L_i, MB_i)
end

freqs = 10 .^ range(0, 9, length=900)

println("\n====================================================================")
println(" MASSIVE BATCH EVALUATION: 600 CIRCUITS (300 DM + 300 CM SIMULTANEOUS)")
println("====================================================================")

results_x_dm = Vector{SVector{length(sim_dm.unknown_nodes), Float64}}(undef, N)
results_ac_dm = Vector{Matrix{ComplexF64}}(undef, N)
results_ac_cm = Vector{Matrix{ComplexF64}}(undef, N)

println(">>> Running 600-Batch DC & AC Solves on Julia Threads...")
start_time = time()

@threads for i in 1:N
    x_dm, err = solve_dc(sim_dm, circuits[i])
    results_x_dm[i] = x_dm
    results_ac_dm[i] = solve_ac(sim_dm, circuits[i], x_dm, freqs)
    
    x_cm, err_cm = solve_dc(sim_cm, circuits[i])
    results_ac_cm[i] = solve_ac(sim_cm, circuits[i], x_cm, freqs)
end

total_time = time() - start_time
println("Julia 600-Circuit Total Time: $(round(total_time, digits=4)) s")

idx_vb_dm = findfirst(==("vb"), sim_dm.unknown_nodes)
idx_vm_dm = findfirst(==("vm"), sim_dm.unknown_nodes)
idx_vT_dm = findfirst(==("vT"), sim_dm.unknown_nodes)
idx_vout_dm = findfirst(==("vout"), sim_dm.unknown_nodes)
idx_vout_cm = findfirst(==("vout"), sim_cm.unknown_nodes)

jl_vb = zeros(N); jl_vm = zeros(N); jl_vT = zeros(N); jl_vout = zeros(N)
jl_AVD = zeros(N); jl_BW = zeros(N); jl_UGF = zeros(N); jl_PM = zeros(N); jl_AVCM = zeros(N)

for i in 1:N
    # Extract exact DC Nodes
    jl_vb[i] = results_x_dm[i][idx_vb_dm]
    jl_vm[i] = results_x_dm[i][idx_vm_dm]
    jl_vT[i] = results_x_dm[i][idx_vT_dm]
    jl_vout[i] = results_x_dm[i][idx_vout_dm]

    gain_dm_db = 20 .* log10.(abs.(results_ac_dm[i][idx_vout_dm, :]))
    gain_cm_db = 20 .* log10.(abs.(results_ac_cm[i][idx_vout_cm, :]))
    phase_dm = unwrap_phase_custom(angle.(results_ac_dm[i][idx_vout_dm, :])) .* 180 ./ pi
    
    jl_AVD[i] = maximum(gain_dm_db)
    jl_AVCM[i] = gain_cm_db[1]
    
    bw_idx = findfirst(x -> x <= jl_AVD[i] - 3.0, gain_dm_db)
    jl_BW[i] = bw_idx !== nothing ? freqs[bw_idx] : freqs[end]
    
    ugf_i = argmin(abs.(gain_dm_db))
    jl_UGF[i] = freqs[ugf_i]
    jl_PM[i] = 180.0 + phase_dm[ugf_i]
end

# ==============================================================================
# 6. LTSPICE VERIFICATION & PLOTTING
# ==============================================================================
println("\n>>> Running SPICE (Sequential) for all 300 points (DC & AC)...")
BASE_DIR = pwd()

spice_vb = zeros(N); spice_vm = zeros(N); spice_vT = zeros(N); spice_vout = zeros(N)
spice_AVD = zeros(N); spice_BW = zeros(N); spice_UGF = zeros(N)
spice_PM = zeros(N); spice_AVCM = zeros(N); spice_CMRR = zeros(N)
spice_traces = Dict{Int, Dict}()

start_spice = time()
for i in 1:N
    if i % 10 == 0 || i == N
        print("\r SPICE Progress: $i/$N... ($(round((i/N)*100, digits=1))%)")
        flush(stdout)
    end
    
    M_i, L_i, MB_i = sampled_combinations[i]
    try
        v_dc = evaluate_5t_ota_dc(M_i, L_i, M_i, L_i, M_i, L_i, MB_i, Ib, CL, Vicm, BASE_DIR)
        spice_vb[i] = pyconvert(Float64, v_dc["vb"])
        spice_vm[i] = pyconvert(Float64, v_dc["vm"])
        spice_vT[i] = pyconvert(Float64, v_dc["vT"])
        spice_vout[i] = pyconvert(Float64, v_dc["vout"])
    
        ac_specs = evaluate_5t_ota_ac(M_i, L_i, M_i, L_i, M_i, L_i, MB_i, Ib, CL, Vicm, BASE_DIR)
        spice_AVD[i], spice_BW[i] = pyconvert(Float64, ac_specs["AVD"]), pyconvert(Float64, ac_specs["BW"])
        spice_UGF[i], spice_PM[i] = pyconvert(Float64, ac_specs["UGF"]), pyconvert(Float64, ac_specs["PM"])
        spice_AVCM[i], spice_CMRR[i] = pyconvert(Float64, ac_specs["AVCM"]), pyconvert(Float64, ac_specs["CMRR"])
        
        # Save exact traces for inline plotting
        spice_traces[i] = Dict(
            "freqs" => pyconvert(Array, ac_specs["freqs"]),
            "mag"   => pyconvert(Array, ac_specs["mag"]),
            "phase" => pyconvert(Array, ac_specs["phase"])
        )
    catch e
        if i == 1
            @error "LTspice failed on first iteration: " exception=e
        end
    end
end
println("\nSPICE evaluation complete!")
spice_time = time() - start_spice
println("SPICE Total Time: $(round(spice_time, digits=4)) s")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")

# Compute Exact DC Node Errors
err_vb = abs.(jl_vb .- spice_vb)
err_vm = abs.(jl_vm .- spice_vm)
err_vT = abs.(jl_vT .- spice_vT)
err_vout = abs.(jl_vout .- spice_vout)

max_err_circuit = max.(err_vb, err_vm, err_vT, err_vout)
worst_idx = sortperm(max_err_circuit, rev=true)[1:10]

# IMPORT MATPLOTLIB FOR JULIA INLINE PLOTTING
plt = pyimport("matplotlib.pyplot")

println("\n======================================================================")
println(" TOP 10 WORST CASE CIRCUITS: JULIA VS LTSPICE (DC & AC)")
println("======================================================================")
for (rank, idx) in enumerate(worst_idx)
    M_i, L_i, MB_i = sampled_combinations[idx]
    
    @printf("Rank %d (Circuit #%d) - Max DC Error: %.3f mV\n", rank, idx-1, max_err_circuit[idx]*1000)
    @printf("Params: M=%d (W=%.2fum), L=%.2fum, MB=%d (WB=%.2fum)\n\n", M_i, M_i*0.5, L_i*1e6, MB_i, MB_i*0.5)
    
    println("  DC Nodes | Julia (V)   | LTspice (V) | Diff (mV)")
    @printf("  vb       | %.4f      | %.4f      | %.3f\n", jl_vb[idx], spice_vb[idx], err_vb[idx]*1000)
    @printf("  vm       | %.4f      | %.4f      | %.3f\n", jl_vm[idx], spice_vm[idx], err_vm[idx]*1000)
    @printf("  vT       | %.4f      | %.4f      | %.3f\n", jl_vT[idx], spice_vT[idx], err_vT[idx]*1000)
    @printf("  vout     | %.4f      | %.4f      | %.3f\n\n", jl_vout[idx], spice_vout[idx], err_vout[idx]*1000)
    
    println("  AC Specs | Julia       | LTspice     | Diff")
    @printf("  AVD      | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx], spice_AVD[idx], abs(jl_AVD[idx] - spice_AVD[idx]))
    diff_bw = abs(jl_BW[idx] - spice_BW[idx]) / (spice_BW[idx] + 1e-9) * 100
    @printf("  BW       | %.2e Hz | %.2e Hz | %.2f %%\n", jl_BW[idx], spice_BW[idx], diff_bw)
    diff_ugf = abs(jl_UGF[idx] - spice_UGF[idx]) / (spice_UGF[idx] + 1e-9) * 100
    @printf("  UGF      | %.2e Hz | %.2e Hz | %.2f %%\n", jl_UGF[idx], spice_UGF[idx], diff_ugf)
    @printf("  PM       | %6.2f deg  | %6.2f deg  | %.2f deg\n", jl_PM[idx], spice_PM[idx], abs(jl_PM[idx] - spice_PM[idx]))
    @printf("  AVCM     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVCM[idx], spice_AVCM[idx], abs(jl_AVCM[idx] - spice_AVCM[idx]))
    @printf("  CMRR     | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx]-jl_AVCM[idx], spice_CMRR[idx], abs((jl_AVD[idx]-jl_AVCM[idx]) - spice_CMRR[idx]))
    
    gain_dm_db_idx = 20 .* log10.(abs.(results_ac_dm[idx][idx_vout_dm, :]))
    phase_dm_idx = unwrap_phase_custom(angle.(results_ac_dm[idx][idx_vout_dm, :])) .* 180 ./ pi
    freqs_array = collect(freqs) 
    
    plt.figure(figsize=(10, 8))
    
    # Magnitude Plot
    plt.subplot(2, 1, 1)
    plt.semilogx(freqs_array, gain_dm_db_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["mag"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.title("Rank $rank (Circuit #$(idx-1)) - AC Response")
    plt.ylabel("Magnitude (dB)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    # Phase Plot
    plt.subplot(2, 1, 2)
    plt.semilogx(freqs_array, phase_dm_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["phase"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.ylabel("Phase (deg)")
    plt.xlabel("Frequency (Hz)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.tight_layout()
    
    # Fix for Jupyter cell hang: saves and displays inline directly 
    plt.savefig("temp_plot_5t.png", bbox_inches="tight")
    display(MIME("image/png"), read("temp_plot_5t.png"))
    plt.close()
    
    println("-" ^ 70)
end
```
