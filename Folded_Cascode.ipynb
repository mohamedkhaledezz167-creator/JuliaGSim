```julia
# ==============================================================================
# 1. Call the LUT definitions and the simulator Engine
# ==============================================================================
ENV["JULIA_PYTHONCALL_EXE"] = "C:\\Users\\mohamed\\.conda\\envs\\Thesis_Optimizer_Sandbox_Copy\\python.exe"
# 1. Load technology model definitions
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Surrogate_definitions.jl")

# 2. Load the Engine (choose Static for small circuits or Sparse for large circuits)
include("C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Julia_GSim\\Vector_Sim_Engine\\Static_Sim.jl")


# ==============================================================================
# 2.Helper to unwrap phase angles natively in Julia 
# ==============================================================================
function unwrap_phase_custom(p)
    p_out = copy(p)
    for j in 2:length(p_out)
        d = p_out[j] - p_out[j-1]
        p_out[j] -= 2pi * round(d / 2pi) 
    end
    return p_out
end

# ==============================================================================
# 3. LTSPICE VERIFICATION GROUND TRUTH 
# ==============================================================================
g = pydict()
pyexec(raw"""
import os, re, subprocess, numpy as np
from PyLTSpice import RawRead

def evaluate_fc_ota_dc_voltages(md, Ld, mp_src, Lp, mss, Lss, mp_csc, mm_mir, mm_csc, Lm, mb, Ib, I_cascode, vicm, BASE_DIR='.'):
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR,"Folded_DC.sp")
    rp = os.path.join(BASE_DIR,"Folded_DC.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA',''),'LTspice','lib','cmp','standard.mos')

    m_bias_p_csc = mp_csc / 4.0
    m_bias_n_csc = mm_csc / 4.0
    I_tail = Ib * (mss/mb)
    I_pmos = I_tail/2.0 + I_cascode

    nl = f'''* FC OTA DC AutoGen
Vdd Vdd 0 3
vicm vcm 0 {vicm}
vinp vip vcm 0
vinn vin vcm 0

IB Vdd vb {Ib}
M_bias vb vb 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}

I_Vbp Vbp 0 {I_pmos}
M_bias_p Vbp Vbp Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

I_VcscP VcscP 0 {I_cascode}
M_bias_cscP VcscP VcscP Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={m_bias_p_csc}

I_VcscN Vdd VcscN {I_cascode}
M_bias_cscN VcscN VcscN 0 0 nmos_3p3 l={Lm} w=0.5u m={m_bias_n_csc}

M_tail vT vb 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M1 vfold_p vip vT 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2 vfold_n vin vT 0 nmos_3p3 l={Ld} w=0.5u m={md}

M3 vfold_p Vbp Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}
M4 vfold_n Vbp Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

M5 vx_left VcscP vfold_p Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}
M6 vout VcscP vfold_n Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}

M7 vy_left vx_left 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}
M8 vout_bot vx_left 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}

M9 vx_left VcscN vy_left 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}
M10 vout VcscN vout_bot 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.op
.end
'''
    with open(sp,'w',encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH,"-b","-run",sp],cwd=BASE_DIR,capture_output=True)
    raw=RawRead(rp); tr=raw.get_trace_names()
    def gv(n):
        for t in tr:
            if t.lower()==n.lower(): return float(raw.get_trace(t).get_wave(0)[0])
        return 0.0
    nodes = ['vb', 'Vbp', 'VcscP', 'VcscN', 'vT', 'vfold_p', 'vfold_n', 'vx_left', 'vy_left', 'vout_bot', 'vout']
    return {k:gv(f"V({k})") for k in nodes}

def evaluate_fc_ota_ac_ltspice(md, Ld, mp_src, Lp, mss, Lss, mp_csc, mm_mir, mm_csc, Lm, mb, Ib, I_cascode, vicm, CL, BASE_DIR='.'):
    LTSPICE_PATH = r"C:\Users\Mohamed\AppData\Local\Programs\ADI\LTspice\LTspice.exe"
    sp = os.path.join(BASE_DIR,"FC_OTA_AC_TEST.sp")
    rp = os.path.join(BASE_DIR,"FC_OTA_AC_TEST.raw")
    lib = os.path.join(os.environ.get('LOCALAPPDATA',''),'LTspice','lib','cmp','standard.mos')
    
    m_bias_p_csc = mp_csc / 4.0
    m_bias_n_csc = mm_csc / 4.0
    I_tail = Ib * (mss/mb)
    I_pmos = I_tail/2.0 + I_cascode
    
    nl = f'''* FC OTA AC AutoGen
Vdd Vdd 0 3

vicm_dm vcm_dm 0 {vicm}
vinp_dm vip_dm vcm_dm AC 0.5
vinn_dm vin_dm vcm_dm AC -0.5

IB_dm Vdd vb_dm {Ib}
M_bias_dm vb_dm vb_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}

I_Vbp_dm Vbp_dm 0 {I_pmos}
M_bias_p_dm Vbp_dm Vbp_dm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

I_VcscP_dm VcscP_dm 0 {I_cascode}
M_bias_cscP_dm VcscP_dm VcscP_dm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={m_bias_p_csc}

I_VcscN_dm Vdd VcscN_dm {I_cascode}
M_bias_cscN_dm VcscN_dm VcscN_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={m_bias_n_csc}

M_tail_dm vT_dm vb_dm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M1_dm vfold_p_dm vip_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2_dm vfold_n_dm vin_dm vT_dm 0 nmos_3p3 l={Ld} w=0.5u m={md}

M3_dm vfold_p_dm Vbp_dm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}
M4_dm vfold_n_dm Vbp_dm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

M5_dm vx_left_dm VcscP_dm vfold_p_dm Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}
M6_dm vout_dm VcscP_dm vfold_n_dm Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}

M7_dm vy_left_dm vx_left_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}
M8_dm vout_bot_dm vx_left_dm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}

M9_dm vx_left_dm VcscN_dm vy_left_dm 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}
M10_dm vout_dm VcscN_dm vout_bot_dm 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}

CL_dm vout_dm 0 {CL}

vicm_ac vicm_ac 0 {vicm} AC 1

IB_cm Vdd vb_cm {Ib}
M_bias_cm vb_cm vb_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mb}

I_Vbp_cm Vbp_cm 0 {I_pmos}
M_bias_p_cm Vbp_cm Vbp_cm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

I_VcscP_cm VcscP_cm 0 {I_cascode}
M_bias_cscP_cm VcscP_cm VcscP_cm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={m_bias_p_csc}

I_VcscN_cm Vdd VcscN_cm {I_cascode}
M_bias_cscN_cm VcscN_cm VcscN_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={m_bias_n_csc}

M_tail_cm vT_cm vb_cm 0 0 nmos_3p3 l={Lss} w=0.5u m={mss}
M1_cm vfold_p_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md}
M2_cm vfold_n_cm vicm_ac vT_cm 0 nmos_3p3 l={Ld} w=0.5u m={md}

M3_cm vfold_p_cm Vbp_cm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}
M4_cm vfold_n_cm Vbp_cm Vdd Vdd pmos_3p3 l={Lp} w=0.5u m={mp_src}

M5_cm vx_left_cm VcscP_cm vfold_p_cm Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}
M6_cm vout_cm VcscP_cm vfold_n_cm Vdd pmos_3p3 l={Lp} w=0.5u m={mp_csc}

M7_cm vy_left_cm vx_left_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}
M8_cm vout_bot_cm vx_left_cm 0 0 nmos_3p3 l={Lm} w=0.5u m={mm_mir}

M9_cm vx_left_cm VcscN_cm vy_left_cm 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}
M10_cm vout_cm VcscN_cm vout_bot_cm 0 nmos_3p3 l={Lm} w=0.5u m={mm_csc}

CL_cm vout_cm 0 {CL}

.model NMOS NMOS
.model PMOS PMOS
.lib {lib}
.inc cmos_tt.txt
.ac dec 100 1 1e9
.end
'''
    with open(sp,'w',encoding='utf-8') as f: f.write(nl)
    subprocess.run([LTSPICE_PATH,"-b","-run",sp],cwd=BASE_DIR,capture_output=True)
    raw   = RawRead(rp)
    freqs = np.abs(raw.get_trace("frequency").get_wave(0))
    
    vout_dm_c = raw.get_trace("V(vout_dm)").get_wave(0)
    vout_cm_c = raw.get_trace("V(vout_cm)").get_wave(0)
    
    dm    = 20*np.log10(np.abs(vout_dm_c))
    cm    = 20*np.log10(np.abs(vout_cm_c))
    phase = np.unwrap(np.angle(vout_dm_c)) * 180 / np.pi
    
    AVD   = np.max(dm)
    bwi   = np.where(dm <= AVD-3.0)[0]
    BW    = freqs[bwi[0]] if len(bwi) else freqs[-1]
    ugf_i = np.argmin(np.abs(dm))
    UGF   = freqs[ugf_i]
    PM    = 180.0 + phase[ugf_i]
    AVCM  = cm[0]
    
    return {'AVD':AVD,'BW':BW,'UGF':UGF,'PM':PM,'AVCM':AVCM,'CMRR':AVD-AVCM,
            'freqs': freqs, 'mag': dm, 'phase': phase}
""", g)
evaluate_fc_ota_dc = g["evaluate_fc_ota_dc_voltages"]
evaluate_fc_ota_ac = g["evaluate_fc_ota_ac_ltspice"]


# ==============================================================================
# 4. BUILDING THE DIFFERENTIAL AND COMMON MODE FOLDED CASCODE OTA
# ==============================================================================
VDD, Vicm = 3.0, 1.5
Ib, I_cascode = 20e-6, 20e-6
CL = 1e-12

sim_dm = JuliaGSim(surrogate_model)
add_vsource!(sim_dm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_dm, "V_VIP", "vip", "0", Vicm)
add_vsource!(sim_dm, "V_VIN", "vin", "0", Vicm)
add_vac_source!(sim_dm, "V_VIP_AC", "vip", "0", 0.5)
add_vac_source!(sim_dm, "V_VIN_AC", "vin", "0", -0.5)

add_isource!(sim_dm, "IB", "VDD", "vb", Ib)
add_isource!(sim_dm, "I_Vbp", "Vbp", "0", :I_pmos_val)
add_isource!(sim_dm, "I_VcscP", "VcscP", "0", I_cascode)
add_isource!(sim_dm, "I_VcscN", "VDD", "VcscN", I_cascode)

add_nmos!(sim_dm, "M_bias", d="vb", g="vb", s="0", b="0", M=:mb, L=:Lss)
add_pmos!(sim_dm, "M_bias_p", d="Vbp", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)
add_pmos!(sim_dm, "M_bias_cscP", d="VcscP", g="VcscP", s="VDD", b="VDD", M=:m_bias_p_csc, L=:Lp)
add_nmos!(sim_dm, "M_bias_cscN", d="VcscN", g="VcscN", s="0", b="0", M=:m_bias_n_csc, L=:Lm)

add_nmos!(sim_dm, "M_tail", d="vT", g="vb", s="0", b="0", M=:mss, L=:Lss)

# SWAPPED vip and vin for correct polarity phase
add_nmos!(sim_dm, "M1", d="vfold_p", g="vip", s="vT", b="0", M=:md, L=:Ld)
add_nmos!(sim_dm, "M2", d="vfold_n", g="vin", s="vT", b="0", M=:md, L=:Ld)

add_pmos!(sim_dm, "M3", d="vfold_p", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)
add_pmos!(sim_dm, "M4", d="vfold_n", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)

add_pmos!(sim_dm, "M5", d="vx_left", g="VcscP", s="vfold_p", b="VDD", M=:mp_csc, L=:Lp)
add_pmos!(sim_dm, "M6", d="vout", g="VcscP", s="vfold_n", b="VDD", M=:mp_csc, L=:Lp)

add_nmos!(sim_dm, "M7", d="vy_left", g="vx_left", s="0", b="0", M=:mm_mir, L=:Lm)
add_nmos!(sim_dm, "M8", d="vout_bot", g="vx_left", s="0", b="0", M=:mm_mir, L=:Lm)
add_nmos!(sim_dm, "M9", d="vx_left", g="VcscN", s="vy_left", b="0", M=:mm_csc, L=:Lm)
add_nmos!(sim_dm, "M10", d="vout", g="VcscN", s="vout_bot", b="0", M=:mm_csc, L=:Lm)

add_cap!(sim_dm, "CL", "vout", "0", CL)
compile!(sim_dm)


sim_cm = JuliaGSim(surrogate_model)
add_vsource!(sim_cm, "V_VDD", "VDD", "0", VDD)
add_vsource!(sim_cm, "VICM", "vicm", "0", Vicm)
add_vac_source!(sim_cm, "VICM_AC", "vicm", "0", 1.0)

add_isource!(sim_cm, "IB", "VDD", "vb", Ib)
add_isource!(sim_cm, "I_Vbp", "Vbp", "0", :I_pmos_val)
add_isource!(sim_cm, "I_VcscP", "VcscP", "0", I_cascode)
add_isource!(sim_cm, "I_VcscN", "VDD", "VcscN", I_cascode)

add_nmos!(sim_cm, "M_bias", d="vb", g="vb", s="0", b="0", M=:mb, L=:Lss)
add_pmos!(sim_cm, "M_bias_p", d="Vbp", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)
add_pmos!(sim_cm, "M_bias_cscP", d="VcscP", g="VcscP", s="VDD", b="VDD", M=:m_bias_p_csc, L=:Lp)
add_nmos!(sim_cm, "M_bias_cscN", d="VcscN", g="VcscN", s="0", b="0", M=:m_bias_n_csc, L=:Lm)

add_nmos!(sim_cm, "M_tail", d="vT", g="vb", s="0", b="0", M=:mss, L=:Lss)
add_nmos!(sim_cm, "M1", d="vfold_p", g="vicm", s="vT", b="0", M=:md, L=:Ld)
add_nmos!(sim_cm, "M2", d="vfold_n", g="vicm", s="vT", b="0", M=:md, L=:Ld)

add_pmos!(sim_cm, "M3", d="vfold_p", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)
add_pmos!(sim_cm, "M4", d="vfold_n", g="Vbp", s="VDD", b="VDD", M=:mp_src, L=:Lp)

add_pmos!(sim_cm, "M5", d="vx_left", g="VcscP", s="vfold_p", b="VDD", M=:mp_csc, L=:Lp)
add_pmos!(sim_cm, "M6", d="vout", g="VcscP", s="vfold_n", b="VDD", M=:mp_csc, L=:Lp)

add_nmos!(sim_cm, "M7", d="vy_left", g="vx_left", s="0", b="0", M=:mm_mir, L=:Lm)
add_nmos!(sim_cm, "M8", d="vout_bot", g="vx_left", s="0", b="0", M=:mm_mir, L=:Lm)
add_nmos!(sim_cm, "M9", d="vx_left", g="VcscN", s="vy_left", b="0", M=:mm_csc, L=:Lm)
add_nmos!(sim_cm, "M10", d="vout", g="VcscN", s="vout_bot", b="0", M=:mm_csc, L=:Lm)

add_cap!(sim_cm, "CL", "vout", "0", CL)
compile!(sim_cm)


# ==============================================================================
# 5. PARALLEL BATCH EVALUATION
# ==============================================================================
struct CircuitParams
    md::Float64; mp_src::Float64; mss::Float64; mp_csc::Float64
    mm_mir::Float64; mm_csc::Float64; mb::Float64
    m_bias_p_csc::Float64; m_bias_n_csc::Float64
    Ld::Float64; Lm::Float64; Lss::Float64; Lp::Float64
    I_pmos_val::Float64
end

W_arr = range(20e-6, stop=40e-6, length=20)
L_arr = range(1e-6, stop=10e-6, length=15)
combinations = vec(collect(Iterators.product(L_arr, W_arr)))
N = length(combinations) # 300

circuits = Vector{CircuitParams}(undef, N)
for i in 1:N
    L_i, W_i = combinations[i]
    
    M_unit = 0.5e-6
    md_i = W_i / M_unit
    mp_src_i = W_i / M_unit
    mss_i = W_i / M_unit
    mp_csc_i = (2.5 * W_i) / M_unit
    mm_mir_i = (0.5 * W_i) / M_unit
    mm_csc_i = (0.5 * W_i) / M_unit
    mb_i = (0.5 * W_i) / M_unit
    
    Ld_i = L_i
    Lm_i = L_i
    Lss_i = L_i
    Lp_i = 0.5 * L_i
    
    m_bias_p_csc_i = mp_csc_i / 4.0
    m_bias_n_csc_i = mm_csc_i / 4.0
    
    I_tail_val = Ib * 2.0 
    I_pmos_val = I_tail_val / 2.0 + I_cascode
    
    circuits[i] = CircuitParams(md_i, mp_src_i, mss_i, mp_csc_i, mm_mir_i, mm_csc_i, mb_i, 
                                m_bias_p_csc_i, m_bias_n_csc_i, Ld_i, Lm_i, Lss_i, Lp_i, I_pmos_val)
end

freqs = 10 .^ range(0, stop=9, length=900)

println("\n====================================================================")
println(" MASSIVE BATCH EVALUATION: $(N*2) CIRCUITS (FOLDED CASCODE DM + CM)")
println("====================================================================")

results_x_dm = Vector{SVector{length(sim_dm.unknown_nodes), Float64}}(undef, N)
results_ac_dm = Vector{Matrix{ComplexF64}}(undef, N)
results_ac_cm = Vector{Matrix{ComplexF64}}(undef, N)

println(">>> Running $(N)-Batch DC & AC Solves on Julia Threads...")
start_time = time()

@threads for i in 1:N
    x_dm, err = solve_dc(sim_dm, circuits[i])
    results_x_dm[i] = x_dm
    results_ac_dm[i] = solve_ac(sim_dm, circuits[i], x_dm, freqs)
    
    x_cm, err_cm = solve_dc(sim_cm, circuits[i])
    results_ac_cm[i] = solve_ac(sim_cm, circuits[i], x_cm, freqs)
end

total_time = time() - start_time
println("Julia 300-Circuit Total Time: $(round(total_time, digits=4)) s")

idx_vb = findfirst(==("vb"), sim_dm.unknown_nodes)
idx_Vbp = findfirst(==("Vbp"), sim_dm.unknown_nodes)
idx_VcscP = findfirst(==("VcscP"), sim_dm.unknown_nodes)
idx_VcscN = findfirst(==("VcscN"), sim_dm.unknown_nodes)
idx_vT = findfirst(==("vT"), sim_dm.unknown_nodes)
idx_vfold_p = findfirst(==("vfold_p"), sim_dm.unknown_nodes)
idx_vfold_n = findfirst(==("vfold_n"), sim_dm.unknown_nodes)
idx_vx_left = findfirst(==("vx_left"), sim_dm.unknown_nodes)
idx_vy_left = findfirst(==("vy_left"), sim_dm.unknown_nodes)
idx_vout_bot = findfirst(==("vout_bot"), sim_dm.unknown_nodes)
idx_vout_dm = findfirst(==("vout"), sim_dm.unknown_nodes)
idx_vout_cm = findfirst(==("vout"), sim_cm.unknown_nodes)

jl_vb = zeros(N); jl_Vbp = zeros(N); jl_VcscP = zeros(N); jl_VcscN = zeros(N)
jl_vT = zeros(N); jl_vfold_p = zeros(N); jl_vfold_n = zeros(N)
jl_vx_left = zeros(N); jl_vy_left = zeros(N); jl_vout_bot = zeros(N); jl_vout = zeros(N)

jl_AVD = zeros(N); jl_BW = zeros(N); jl_UGF = zeros(N); jl_PM = zeros(N); jl_AVCM = zeros(N)

for i in 1:N
    jl_vb[i] = results_x_dm[i][idx_vb]
    jl_Vbp[i] = results_x_dm[i][idx_Vbp]
    jl_VcscP[i] = results_x_dm[i][idx_VcscP]
    jl_VcscN[i] = results_x_dm[i][idx_VcscN]
    jl_vT[i] = results_x_dm[i][idx_vT]
    jl_vfold_p[i] = results_x_dm[i][idx_vfold_p]
    jl_vfold_n[i] = results_x_dm[i][idx_vfold_n]
    jl_vx_left[i] = results_x_dm[i][idx_vx_left]
    jl_vy_left[i] = results_x_dm[i][idx_vy_left]
    jl_vout_bot[i] = results_x_dm[i][idx_vout_bot]
    jl_vout[i] = results_x_dm[i][idx_vout_dm]

    gain_dm_db = 20 .* log10.(abs.(results_ac_dm[i][idx_vout_dm, :]))
    gain_cm_db = 20 .* log10.(abs.(results_ac_cm[i][idx_vout_cm, :]))
    phase_dm = unwrap_phase_custom(angle.(results_ac_dm[i][idx_vout_dm, :])) .* 180 ./ pi
    
    jl_AVD[i] = maximum(gain_dm_db)
    jl_AVCM[i] = gain_cm_db[1]
    
    bw_idx = findfirst(x -> x <= jl_AVD[i] - 3.0, gain_dm_db)
    jl_BW[i] = bw_idx !== nothing ? freqs[bw_idx] : freqs[end]
    
    ugf_i = argmin(abs.(gain_dm_db))
    jl_UGF[i] = freqs[ugf_i]
    jl_PM[i] = 180.0 + phase_dm[ugf_i]
end

# ==============================================================================
# 6. LTSPICE VERIFICATION & PLOTTING
# ==============================================================================
println("\n>>> Running SPICE (Sequential) for all 300 points (DC & AC)...")
BASE_DIR = "C:\\Users\\mohamed\\Desktop\\ML\\CAD_projects\\Netlists\\GF_180nm_tech_work\\project\\New_Sim_tests\\Allah_help_me2\\Allah_Save_me\\New folder\\system\\Testing\\Spice_vs_Julia_Benchmarking"

spice_vb = zeros(N); spice_Vbp = zeros(N); spice_VcscP = zeros(N); spice_VcscN = zeros(N)
spice_vT = zeros(N); spice_vfold_p = zeros(N); spice_vfold_n = zeros(N)
spice_vx_left = zeros(N); spice_vy_left = zeros(N); spice_vout_bot = zeros(N); spice_vout = zeros(N)

spice_AVD = zeros(N); spice_BW = zeros(N); spice_UGF = zeros(N)
spice_PM = zeros(N); spice_AVCM = zeros(N); spice_CMRR = zeros(N)
spice_traces = Dict{Int, Dict}()

start_spice = time()
for i in 1:N
    if i % 10 == 0 || i == N
        print("\r SPICE Progress: $i/$N... ($(round((i/N)*100, digits=1))%)")
        flush(stdout)
    end
    
    try
        v_dc = evaluate_fc_ota_dc(circuits[i].md, circuits[i].Ld, circuits[i].mp_src, circuits[i].Lp, 
                                  circuits[i].mss, circuits[i].Lss, circuits[i].mp_csc, circuits[i].mm_mir, 
                                  circuits[i].mm_csc, circuits[i].Lm, circuits[i].mb, Ib, I_cascode, Vicm, BASE_DIR)
        spice_vb[i] = pyconvert(Float64, v_dc["vb"])
        spice_Vbp[i] = pyconvert(Float64, v_dc["Vbp"])
        spice_VcscP[i] = pyconvert(Float64, v_dc["VcscP"])
        spice_VcscN[i] = pyconvert(Float64, v_dc["VcscN"])
        spice_vT[i] = pyconvert(Float64, v_dc["vT"])
        spice_vfold_p[i] = pyconvert(Float64, v_dc["vfold_p"])
        spice_vfold_n[i] = pyconvert(Float64, v_dc["vfold_n"])
        spice_vx_left[i] = pyconvert(Float64, v_dc["vx_left"])
        spice_vy_left[i] = pyconvert(Float64, v_dc["vy_left"])
        spice_vout_bot[i] = pyconvert(Float64, v_dc["vout_bot"])
        spice_vout[i] = pyconvert(Float64, v_dc["vout"])
    
        ac_specs = evaluate_fc_ota_ac(circuits[i].md, circuits[i].Ld, circuits[i].mp_src, circuits[i].Lp, 
                                      circuits[i].mss, circuits[i].Lss, circuits[i].mp_csc, circuits[i].mm_mir, 
                                      circuits[i].mm_csc, circuits[i].Lm, circuits[i].mb, Ib, I_cascode, Vicm, CL, BASE_DIR)
        spice_AVD[i], spice_BW[i] = pyconvert(Float64, ac_specs["AVD"]), pyconvert(Float64, ac_specs["BW"])
        spice_UGF[i], spice_PM[i] = pyconvert(Float64, ac_specs["UGF"]), pyconvert(Float64, ac_specs["PM"])
        spice_AVCM[i], spice_CMRR[i] = pyconvert(Float64, ac_specs["AVCM"]), pyconvert(Float64, ac_specs["CMRR"])
        
        spice_traces[i] = Dict(
            "freqs" => pyconvert(Array, ac_specs["freqs"]),
            "mag"   => pyconvert(Array, ac_specs["mag"]),
            "phase" => pyconvert(Array, ac_specs["phase"])
        )
    catch
    end
end
println("\nSPICE evaluation complete!")
spice_time = time() - start_spice
println("SPICE Total Time: $(round(spice_time, digits=4)) s")

speedup = spice_time / total_time
println("\n======================================================================")
println(" 🚀 PERFORMANCE SUMMARY")
println("======================================================================")
@printf(" JuliaGSim Time : %.3f seconds\n", total_time)
@printf(" LTspice Time   : %.3f seconds\n", spice_time)
@printf(" Speedup Factor : %.2fx faster than LTspice!\n", speedup)
println("======================================================================\n")


err_vb = abs.(jl_vb .- spice_vb)
err_Vbp = abs.(jl_Vbp .- spice_Vbp)
err_VcscP = abs.(jl_VcscP .- spice_VcscP)
err_VcscN = abs.(jl_VcscN .- spice_VcscN)
err_vT = abs.(jl_vT .- spice_vT)
err_vfold_p = abs.(jl_vfold_p .- spice_vfold_p)
err_vfold_n = abs.(jl_vfold_n .- spice_vfold_n)
err_vx_left = abs.(jl_vx_left .- spice_vx_left)
err_vy_left = abs.(jl_vy_left .- spice_vy_left)
err_vout_bot = abs.(jl_vout_bot .- spice_vout_bot)
err_vout = abs.(jl_vout .- spice_vout)

max_err_circuit = max.(err_vb, err_Vbp, err_VcscP, err_VcscN, err_vT, err_vfold_p, err_vfold_n, err_vx_left, err_vy_left, err_vout_bot, err_vout)
worst_idx = sortperm(max_err_circuit, rev=true)[1:10]

plt = pyimport("matplotlib.pyplot")

println("\n======================================================================")
println(" TOP 10 WORST CASE CIRCUITS: JULIA VS LTSPICE (DC & AC)")
println("======================================================================")
for (rank, idx) in enumerate(worst_idx)
    md_i = circuits[idx].md
    L_i = circuits[idx].Ld
    
    W_i = md_i * 0.5e-6
    
    @printf("Rank %d (Circuit #%d) - Max DC Error: %.3f mV\n", rank, idx-1, max_err_circuit[idx]*1000)
    @printf("Params: W=%.2fum, L=%.2fum\n\n", W_i*1e6, L_i*1e6)
    
    println("  DC Nodes  | Julia (V)   | LTspice (V) | Diff (mV)")
    @printf("  vb        | %.4f      | %.4f      | %.3f\n", jl_vb[idx], spice_vb[idx], err_vb[idx]*1000)
    @printf("  Vbp       | %.4f      | %.4f      | %.3f\n", jl_Vbp[idx], spice_Vbp[idx], err_Vbp[idx]*1000)
    @printf("  VcscP     | %.4f      | %.4f      | %.3f\n", jl_VcscP[idx], spice_VcscP[idx], err_VcscP[idx]*1000)
    @printf("  VcscN     | %.4f      | %.4f      | %.3f\n", jl_VcscN[idx], spice_VcscN[idx], err_VcscN[idx]*1000)
    @printf("  vT        | %.4f      | %.4f      | %.3f\n", jl_vT[idx], spice_vT[idx], err_vT[idx]*1000)
    @printf("  vfold_p   | %.4f      | %.4f      | %.3f\n", jl_vfold_p[idx], spice_vfold_p[idx], err_vfold_p[idx]*1000)
    @printf("  vfold_n   | %.4f      | %.4f      | %.3f\n", jl_vfold_n[idx], spice_vfold_n[idx], err_vfold_n[idx]*1000)
    @printf("  vx_left   | %.4f      | %.4f      | %.3f\n", jl_vx_left[idx], spice_vx_left[idx], err_vx_left[idx]*1000)
    @printf("  vy_left   | %.4f      | %.4f      | %.3f\n", jl_vy_left[idx], spice_vy_left[idx], err_vy_left[idx]*1000)
    @printf("  vout_bot  | %.4f      | %.4f      | %.3f\n", jl_vout_bot[idx], spice_vout_bot[idx], err_vout_bot[idx]*1000)
    @printf("  vout      | %.4f      | %.4f      | %.3f\n\n", jl_vout[idx], spice_vout[idx], err_vout[idx]*1000)
    
    println("  AC Specs  | Julia       | LTspice     | Diff")
    @printf("  AVD       | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx], spice_AVD[idx], abs(jl_AVD[idx] - spice_AVD[idx]))
    diff_bw = abs(jl_BW[idx] - spice_BW[idx]) / (spice_BW[idx] + 1e-12) * 100
    @printf("  BW        | %.2e Hz | %.2e Hz | %.2f %%\n", jl_BW[idx], spice_BW[idx], diff_bw)
    diff_ugf = abs(jl_UGF[idx] - spice_UGF[idx]) / (spice_UGF[idx] + 1e-12) * 100
    @printf("  UGF       | %.2e Hz | %.2e Hz | %.2f %%\n", jl_UGF[idx], spice_UGF[idx], diff_ugf)
    @printf("  PM        | %6.2f deg  | %6.2f deg  | %.2f deg\n", jl_PM[idx], spice_PM[idx], abs(jl_PM[idx] - spice_PM[idx]))
    @printf("  AVCM      | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVCM[idx], spice_AVCM[idx], abs(jl_AVCM[idx] - spice_AVCM[idx]))
    @printf("  CMRR      | %6.2f dB   | %6.2f dB   | %.2f dB\n", jl_AVD[idx]-jl_AVCM[idx], spice_CMRR[idx], abs((jl_AVD[idx]-jl_AVCM[idx]) - spice_CMRR[idx]))
    
    gain_dm_db_idx = 20 .* log10.(abs.(results_ac_dm[idx][idx_vout_dm, :]))
    phase_dm_idx = unwrap_phase_custom(angle.(results_ac_dm[idx][idx_vout_dm, :])) .* 180 ./ pi
    freqs_array = collect(freqs) 
    
    plt.figure(figsize=(10, 8))
    
    plt.subplot(2, 1, 1)
    plt.semilogx(freqs_array, gain_dm_db_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["mag"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.title("Rank $rank (Circuit #$(idx-1)) - AC Response")
    plt.ylabel("Magnitude (dB)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.subplot(2, 1, 2)
    plt.semilogx(freqs_array, phase_dm_idx, label="JuliaGSim", color="red", linewidth=2)
    if haskey(spice_traces, idx)
        plt.semilogx(spice_traces[idx]["freqs"], spice_traces[idx]["phase"], 
                     label="LTspice", color="blue", linestyle="dashed", linewidth=2)
    end
    plt.ylabel("Phase (deg)")
    plt.xlabel("Frequency (Hz)")
    plt.grid(true, which="both", ls=":")
    plt.legend()
    
    plt.tight_layout()
    plt.savefig("temp_plot_fc.png", bbox_inches="tight")
    display(MIME("image/png"), read("temp_plot_fc.png"))
    plt.close()
    
    println("-" ^ 70)
end
```
